## [Introduction to Claude Artifacts](https://accenture.percipio.com/courses/f140b9f4-f97a-4e9e-99e1-e25d282bcd25)

### Key Takeaway

Claude Artifacts represent a shift from traditional linear chat consultation to **side-by-side, real-time collaboration**. They allow users to generate, iterate on, download, and publish self-contained assets—including Markdown documents, SVG graphics, Mermaid diagrams, HTML sites, and React components—in a dedicated window alongside the chat interface.

---

### 1. Core Concepts & Foundations

- **Definition:** Artifacts are substantial, self-contained pieces of content (documents, code, web components) rendered in a dedicated pane to the right of the conversation window.
- **Enabling Model:** Launched alongside **Claude 3.5 Sonnet** (released June 20, 2024), which benchmarks ahead of GPT-4o and Gemini 1.5 Pro.
- **Interaction Paradigm:** Moves from user-model conversational consultation to iterative co-creation. Artifacts can be updated iteratively via follow-up prompts, downloaded locally, or published via a public URL.

---

### 2. Supported Artifact Types & Use Cases

- **Document Artifacts (Markdown):**
  - Formatted in Markdown (`.md`).
  - Used for brainstorming summaries, procedural documentation, style guides, and formal communications.
  - Can be copied directly to clipboard or downloaded as a file.
- **SVG Artifacts (Scalable Vector Graphics):**
  - Vector code rendered in HTML; lightweight and resizable without loss of quality.
  - Used for icons, web graphics, and simple illustrations.
  - Supports iterative prompt editing (e.g., changing line art to colored fills).
- **Mermaid.js Diagrams:**
  - Written in plain-text Mermaid markup and rendered visually.
  - Used for flowcharts, sales funnels, system architecture, and mind maps.
  - UI includes built-in controls for zoom-in, zoom-out, and full-screen view.
- **HTML Artifacts:**
  - Single-file web pages combining HTML, CSS, and vanilla JavaScript.
  - Can convert unstructured files (e.g., PDF resumes) into styled web pages or run browser-based simulations (e.g., solar system animations).
- **React.js Artifacts:**
  - Standalone React components (`.jsx`/`.tsx`) running live in the Artifact pane.
  - Used for stateful, interactive UIs such as dynamic calculators, recipe serving scalers, and interactive analytics dashboards.
- **Code Artifacts:**
  - Raw code snippets across multiple programming languages.

---

### 3. Comparison: HTML vs. React.js Artifacts

| Feature / Dimension | HTML Artifacts | React.js Artifacts |
|---|---|---|
| **Learning Curve** | Low; readable and easy to troubleshoot | Moderate to High; requires component-based understanding |
| **Execution** | Self-contained single file; runs directly in any browser | Runs inside Claude; requires a React toolchain/codebase to use externally |
| **Interactivity Ceiling** | Good for static layouts and simple JS scripts | High; built for complex state management and dynamic UI updates |
| **Primary Use Cases** | Personal portfolio sites, basic simulations, quick pages | Dynamic applets, multi-view dashboards, data visualizers |

---

### 4. Pre-Installed Libraries in React Artifacts

React Artifacts come pre-configured with four major libraries:

- **shadcn/ui:** Pre-built, accessible UI component primitives.
- **Tailwind CSS:** Utility-first CSS framework for rapid styling.
- **Recharts.js:** Composable, interactive data visualization library (supporting Bar, Line, Area, and Pie charts).
- **Lucide Icons:** Versatile, open-source icon suite.

---

### 5. Sharing Lifecycle: Publishing vs. Remixing

| Attribute | Publishing Artifacts | Remixing Artifacts |
|---|---|---|
| **Action** | Generates a shareable, public web link | Forks an existing published Artifact into a new live chat session |
| **Trigger** | "Publish" button (bottom right of Artifact window) | "Remix Artifact" button on a published Artifact page |
| **Plan Availability** | **Personal plans only** (Claude Free and Claude Pro) | **All plans** (including Claude for Work) |
| **Enterprise / Work Status** | Disabled on Claude for Work for corporate data security | Fully supported across all memberships |
| **Claude Behavior** | Prompts user confirmation against sharing confidential data | Reproduces original Artifact and suggests **3 remix ideas** |

---

### 6. High-Yield Exam Watch-Outs

- **Plan Restriction Asymmetry:** Remember that **Publishing** is blocked on *Claude for Work* accounts for security reasons, while **Remixing** is available to *all* account tiers.
- **Model Attribution:** Artifacts were introduced specifically with **Claude 3.5 Sonnet** on **June 20, 2024**.
- **External Portability:** HTML artifacts can be opened directly as `.html` files in any browser; React components must be integrated into an existing React build setup if exported outside Claude.

## [Introduction to Claude Analysis](https://accenture.percipio.com/courses/35e35ad5-bfc5-48b5-84e6-41538d3e005d)

### Key Takeaway

The **Claude Analysis tool** is a built-in code interpreter that allows Claude to execute **JavaScript code** directly on uploaded CSV files. By translating natural language queries into programmatic operations, the tool automates complex data exploration, cleaning, and statistical analysis, which can then be seamlessly converted into interactive React-based visualization artifacts.

---

### 1. Core Technical Foundations

- **Data Format:** Optimally works with **CSV (Comma-Separated Values)** files, which serve as the universal data exchange format between spreadsheets (Excel, Google Sheets) and data science environments.
- **Execution Environment:** Claude processes data by programmatically generating and running **JavaScript** code behind the scenes. It captures output logs in a virtual terminal console to derive highly accurate analytical answers.
- **Setup Requirement:** To access these code execution features, users must toggle the **Analysis tool** to **On** inside Claude's **Feature Preview** settings (accessible via the user profile menu on the sidebar).

---

### 2. The Data Lifecycle with Claude

The course outlines a structured, four-stage workflow for processing raw data:

- **Stage 1: Preparation & Upload**
  - Export structured sheets from Microsoft Excel or Google Sheets as `.csv` files.
  - Upload the file to Claude using the attachment/document upload utility.
- **Stage 2: Cleaning & Preprocessing (Data Hygiene)**
  - **Inspection:** Claude runs diagnostic code to flag missing values, format disparities, duplicate rows, and metric errors.
  - **Missing Data Handling:** Claude can either remove incomplete rows or perform data imputation (substituting missing fields with statistically neutral estimates).
  - **Normalization:** Standardizes inconsistent date formats, fixes string irregularities (e.g., driver names), and unifies varying measurement units.
- **Stage 3: Feature Engineering & Exploratory Analysis**
  - **Feature Engineering:** Programmatic creation of new, derived columns to surface deeper row-level relationships.
  - **Outlier Detection:** Sweeping datasets to highlight anomalies, statistical outliers, and potential integrity flags.
  - **Summary Statistics:** Automatic extraction of mean, median, mode, ranges, and multi-variable correlations.
- **Stage 4: Visualization (React Artifacts)**
  - Transforms processed CSV data into front-end visual displays.
  - Relies on the pre-installed **Recharts.js** library to build interactive charts (Bar, Line, Area, and Pie).
  - Offers dynamic parameters, such as dropdown menus to toggle chart types on the fly.

---

### 3. Key Analytical Capabilities

- **Pattern & Trend Spotting:** Summarizes broad datasets, identifies growth trajectories, correlates variables (e.g., scale and stability), and uncovers industry concentrations.
- **Stakeholder Recommendations:** Synthesizes multi-step reasoning backed strictly by programmatic computations to deliver targeted business decisions (e.g., recommending investment in high-performing sub-sectors like medical equipment).
- **Conversational Iteration:** Allows users to continuously refine, filter, and drill down into data patterns using successive natural language prompts.

---

### 4. High-Yield Exam Watch-Outs

- **Underlying Language:** Unlike many Python-centric code environments, Claude's internal analysis tool runs **JavaScript** code to inspect, clean, and process uploaded files.
- **Visual Library:** Data visualizations rendered within the React Artifact window are built specifically using the **Recharts** library.
- **Feature Access:** Accessing the analysis workspace requires manual activation of the **Analysis tool** toggle in the **Feature Preview** menu.
- **Plan Restriction Policy:** While personal accounts (Free and Pro) can publish interactive data dashboards to a public URL, publishing is restricted on **team and enterprise accounts** for data privacy and security.

## [Introduction to Claude Projects](https://accenture.percipio.com/courses/f3915463-f455-427d-b516-1196c38e939c)

### Key Takeaway

**Claude Projects** are persistent, customized workspaces within the Claude UI that act as natural language software applications. They bundle together **system prompts** (Project Instructions), **persistent knowledge bases** (Project Knowledge), and **chat histories** to deliver highly context-aware, specialized, and repeatable workflows across multiple interactive sessions.

---

### 1. Core Architecture of a Project

A Claude Project consists of three fundamental, co-operating pillars:

* **Project Instructions (System Prompts):** 
  - Define the overarching persona, operational boundaries, formatting requirements, and structured execution steps.
  - The model references these instructions automatically at the start of every new chat session within that project.
* **Project Knowledge (Persistent Storage):**
  - A dedicated repository for reference files. Supported formats include:
    - **Documents & Text:** PDFs, DOCX, TXT, and Markdown (`.md`).
    - **Structured Data:** CSV spreadsheets and JSON files (fully compatible with Claude's built-in Analysis tool).
    - **Source Code:** Code files across a wide range of extensions.
- **Chat History:**
  - Maintains a persistent log of separate conversations within the workspace for easy reference and retrieval.

---

### 2. High-Value Workspace Workflows

The course highlights three practical archetypes of natural language applications built on Claude Projects:

#### Archetype A: The Interactive Feedback Loop (Writing Coach)
- **Concept:** A multi-document-backed coach that reviews drafts against complex rules and guides the user through revisions rather than rewriting immediately.
- **Project Knowledge:** Uploaded reference standards (e.g., *Chicago Manual of Style*, academic writing guides, internal corporate policies).
- **Execution:** The user submits a rough draft. Claude analyzes the draft against the knowledge base and provides citation-rich feedback. Once the user approves revisions, Claude outputs the final draft as a clean **Markdown Artifact**.

#### Archetype B: The Accumulative Context Engine (Career Coach)
- **Concept:** A project that dynamically builds context and "memorizes" history over weeks or months by saving its own outputs back into its knowledge base.
- **Project Knowledge:** Static files (resume, professional goals, coworker relationship maps) paired with dynamic **Session Logs** (Markdown files of previous coaching conversations).
- **The Accumulative Loop:** 
  1. The user inputs a weekly update or journal entry.
  2. Claude provides tailored guidance drawing on background documents and previous logs.
  3. Claude compiles a session summary as a **Markdown Artifact**.
  4. The user clicks the **"Add to current project" (plus icon)** button in the bottom right.
  5. The Artifact is instantly committed to **Project Knowledge**, ensuring Claude is pre-briefed for the next session.

#### Archetype C: Multi-Step Business Intelligence Dashboards (Data Dashboard)
- **Concept:** Combining multiple Claude features (Data Analysis tool + React Artifacts + Persistent Context) to build an interactive corporate reporting application.
- **Project Knowledge:** Style guides (TechCraft Dataviz Standards), cleaning standards (TechCraft Data Cleaning Standards), and historical context (Previous Years Business Context).
- **Step-by-Step Programmatic Sequence:**
  1. **Upload & Auto-Clean:** The user uploads a fresh `sales.csv`. Claude runs JavaScript to normalize values (e.g., ISO dates, standardized product codes) per the cleaning standards, handling errors recursively.
  2. **Interactive Visualization:** Claude outputs a highly styled **React Artifact** featuring charts tailored to corporate brand colors (using *shadcn/ui*, *Tailwind*, and *Recharts*).
  3. **Contextual Trend Analysis:** Claude writes a narrative report evaluating current performance against historical context stored in the project.
  4. **Executive Infographic:** Claude compiles key insights into a high-level **Executive Summary React Artifact** for leadership.

---

### 3. Feature Matrix & Operations

- **Creation Path:** Sidebar -> **Projects** -> **Create a new project**.
- **Collaborative Capabilities:** On the **Claude for Teams** plan, projects can be shared with team members with customized permissions, allowing collective access to standard system prompts and shared knowledge repositories.
- **The "Add to Project" Utility:** Located in the bottom-right corner of completed Artifacts (displays a plus-sign folder icon). When clicked, it turns blue and appends the Artifact directly to the current project's knowledge base.

---

### 4. High-Yield Exam Watch-Outs

- **The Shared Desk Metaphor:** Understand that a project functions like a specialized desk where all system instructions, background books (knowledge), and past files (chat history) remain permanently organized.
- **Collaborative Permissions:** Project sharing and collaborative workspace permissions are exclusive to **Claude for Teams** accounts.
- **Compounding Context:** Know how the Career Coach example works—specifically, that adding completed chat-log Artifacts back to the Project Knowledge allows the workspace to bypass standard context-window limitations by building a chronological library of past interactions.
- **Error Recovery:** In multi-step data pipelines (like the TechCraft Data Dashboard), Claude is capable of identifying execution errors in its own JavaScript analysis code, correcting the code, and re-running it autonomously before prompting the user for permission to move to the next phase.
