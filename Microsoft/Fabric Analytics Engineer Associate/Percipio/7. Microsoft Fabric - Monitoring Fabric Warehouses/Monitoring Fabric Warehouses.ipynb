## **5. Dynamic Management Views (DMVs)**
### `sys.dm_exec_connections`
Shows:
* Connection information
* Authentication details
* Read/write counts
* Encryption status
### `sys.dm_exec_sessions`
Shows:
* Session details
* Login time
* User name
* Session status
### `sys.dm_exec_requests`
Shows:
* Active requests
* CPU consumption
* Start time
* Request status
### Exam Tip:
| DMV | Purpose |
| --- | --- |
| dm_exec_connections | Connection monitoring |
| dm_exec_sessions | Session monitoring |
| dm_exec_requests | Request/query monitoring |

## **6. Joining DMVs**
Useful joins:
### Connections + Sessions
```sql
SELECT
    connections.connection_id,
    sessions.session_id,
    sessions.login_name
FROM sys.dm_exec_connections connections
INNER JOIN sys.dm_exec_sessions sessions
ON connections.most_recent_session_id = sessions.session_id;
```
Provides:
* User activity
* Login information
* Connection status
### Requests + Sessions
Provides:
* Running queries
* SQL text
* Execution time
* CPU usage

Useful for troubleshooting and identifying bottlenecks.

## **7. Query Insights**
Stores 30 days of query history.
### Views
#### `queryinsights.exec_requests_history`
Historical request information
#### `queryinsights.exec_sessions_history`
Historical session information
#### `queryinsights.long_running_queries`
Tracks queries with high execution times
#### `queryinsights.frequently_run_queries`
Tracks repeatedly executed queries
### Example:
```sql
SELECT *
FROM queryinsights.exec_requests_history
WHERE start_time >= DATEADD(HOUR,-100,GETUTCDATE());
```
Used for:
* Performance trend analysis
* Finding performance bottlenecks
* Identifying frequent queries
* Monitoring user activity
