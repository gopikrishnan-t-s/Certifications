## **2. Microsoft Fabric Administration**
### Explanation
Fabric Administration involves managing:
* Users
* Workspaces
* Capacities
* Security
* Governance
* Tenant Settings
### Key Administrative Areas
#### Tenant Administration
Controls organization-wide settings.
##### Examples:
* Fabric feature enablement
* Workspace creation policies
* Sharing policies
#### Capacity Administration
Manages:
* Capacity Units (CU)
* Resource allocation
* Performance monitoring
#### Workspace Administration
Controls:
* Access
* Collaboration
* Content management
### Summary
Fabric Administration ensures governance, security, and effective resource management across the organization's Fabric environment.

## **3. Fabric Capacities**
### Explanation
A Fabric Capacity provides dedicated compute resources.
### Components
#### Capacity Units (CUs)
Measure compute capacity.
##### Examples:
* F2
* F4
* F8
* F64
#### Capacity Usage
Determines:
* Query performance
* Refresh performance
* Workload concurrency
### Important Concepts
#### Overages
Occur when workloads exceed available capacity.
#### Throttling
Fabric slows or rejects workloads if excessive capacity is consumed.
### Summary
Capacity management is essential for maintaining workload performance and avoiding throttling.

## **7. T-SQL Fundamentals**
### Explanation
T-SQL remains the primary query language for Fabric Warehouses.
### Core Categories
#### DDL (Data Definition Language)
Examples:
```sql
CREATE TABLE
ALTER TABLE
DROP TABLE
```
Used for schema management.
#### DML (Data Manipulation Language)
Examples:
```sql
INSERT
UPDATE
DELETE
```
Used for modifying data.
#### Query Language
Examples:
```sql
SELECT
WHERE
GROUP BY
ORDER BY
```
Used for analytics and reporting.
### Summary
Understanding T-SQL is important for querying, transforming, and managing warehouse data.

## **8. Data Profiling**
### Explanation
Data profiling examines data quality and characteristics.
### Common Statistics
#### Completeness
Checks missing values.
#### Uniqueness
Checks duplicates.
#### Validity
Checks format compliance.
#### Consistency
Checks data integrity.
### Examples
* Null detection
* Duplicate identification
* Invalid data types
* Outlier detection
### Summary
Data profiling helps identify quality issues before data is used for analytics.

## **9. Data Quality Assessment**
### Explanation
Before building reports or semantic models, data quality should be validated.
### Common Quality Issues
#### Missing Values
```
NULL
Blank
```
#### Duplicate Records
Multiple entries for the same business entity.
#### Inconsistent Formats
Examples:
```
2026-08-21
21/08/2026
08-21-2026
```
#### Outliers
Unexpected values that may distort analytics.
### Summary
Data quality directly impacts business reporting accuracy and trust in analytics solutions.

## **10. Data Analytics Best Practices**
### Explanation
Fabric analytics solutions should follow proven design principles.
### Best Practices
#### Use Star Schema
* Better performance
* Easier reporting
* Simpler maintenance
#### Create Semantic Models
* Reusable business layer
* Consistent metrics
#### Optimize Data Refresh
* Incremental refresh
* Direct Lake where appropriate
#### Data Governance
* Apply security policies
* Define ownership
* Monitor usage
### Summary
Good analytics solutions balance performance, governance, scalability, and usability.
