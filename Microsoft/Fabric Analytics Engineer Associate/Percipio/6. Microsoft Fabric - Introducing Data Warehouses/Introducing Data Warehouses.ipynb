## **2. Data Lakehouses and Data Warehouses**
### Explanation
The course explains the three major storage options in Fabric:
### Lakehouse
* Supports structured, semi-structured, and unstructured data.
* Optimized for Spark workloads.
* Uses Delta Lake.
* Supports AI/ML and Big Data processing.
* SQL Analytics Endpoint is read-only.
### Warehouse
* Supports structured relational data only.
* Full T-SQL support.
* Supports ACID transactions.
* Optimized for reporting and analytics.
### Key Difference
| Feature | Lakehouse | Warehouse |
| --- | --- | --- |
| Structured Data | ✅ | ✅ |
| Unstructured Data | ✅ | ❌ |
| Spark | ✅ | ✅ |
| Full T-SQL | ❌ | ✅ |
| Transactions | Limited | ✅ |
| AI/ML | ✅ | Limited |
### Summary
Lakehouses prioritize flexibility and Spark-based analytics, while Warehouses prioritize structured data and SQL-based analytics.

## **3. Power BI Datamarts and KQL Databases**
### Explanation
#### Power BI Datamarts
Used for:
* Self-service analytics
* Business users
* Small-to-medium datasets (up to ~100GB)
* No-code / low-code analytics solutions.
#### KQL Databases
Built for:
* Real-time analytics
* Logs
* Telemetry
* IoT data
* Time-series workloads

Uses:
* Kusto Query Language (KQL)
### Summary
| Storage Type | Best For |
| --- | --- |
| Datamart | Self-service BI |
| Warehouse | Structured Analytics |
| Lakehouse | Big Data & AI |
| KQL Database | Real-Time Analytics |

## **4. Choosing Between Lakehouses and Warehouses**
### Explanation
#### Choose Lakehouse When:
* Working with mixed data types
* Using Spark
* Building ML models
* Data is evolving and flexible
#### Choose Warehouse When:
* Data is fully structured
* Team mainly uses SQL
* Need transactions
* Need advanced SQL objects such as:
    * Stored Procedures
    * Triggers
    * Views
### DP-600 Exam Tip
* Spark-centric → Lakehouse
* T-SQL-centric → Warehouse
### Summary
The decision mainly depends on data type, development skills, and transaction requirements.

## **9. Using COPY INTO to Bulk Load Data**
### Explanation
Primary warehouse ingestion command:
```sql
COPY INTO
```
Loads data from:
* Azure Blob Storage
* ADLS Gen2

Into:
* Warehouse Tables
### Important Parameters
#### FILE_TYPE
```sql
FILE_TYPE='CSV'
```
#### FIRSTROW
```sql
FIRSTROW=2
```

#### CREDENTIAL
Uses SAS Token authentication.
### Summary
COPY INTO is the most important warehouse ingestion command for DP-600.

## **10. Tolerating Errors in Bulk Loads**
### Explanation
#### MAXERRORS
```sql
MAXERRORS = 1
```
Allows a limited number of load failures.
#### ERRORFILE
```sql
ERRORFILE='path'
```
Stores failed rows and error details for investigation.
### Benefits
* Fault tolerance
* Data quality troubleshooting
* Loading continues despite minor issues
### Summary
COPY INTO can gracefully handle and log ingestion errors.

## **12. Performing DDL and DML Operations**
### Explanation
#### DDL (Data Definition Language)
##### CREATE TABLE
```sql
CREATE TABLE
```
##### CTAS
```sql
CREATE TABLE AS SELECT
```
Creates and loads data simultaneously.
#### DML (Data Manipulation Language)
##### INSERT
```sql
INSERT INTO
```
#### UPDATE
```sql
UPDATE
```
#### DELETE
```sql
DELETE
```
### Transactions
Supported in warehouses:
```sql
BEGIN TRANSACTION
COMMIT
ROLLBACK
```
With:
```sql
TRY
CATCH
```
Error handling.
### Summary
Unlike Lakehouses, Warehouses support full DDL, DML, and transaction processing.

## **13. SQL Queries Supported in a Fabric Warehouse**
### Explanation
Comparison of Warehouse and Lakehouse SQL capabilities.
### Warehouse
Supports:
* SELECT
* INSERT
* UPDATE
* DELETE
* CREATE TABLE
* Transactions
### Lakehouse SQL Analytics Endpoint
Supports:
* SELECT
* GROUP BY
* Aggregations

Does NOT support:
* INSERT
* UPDATE
* DELETE
* CREATE TABLE
* Transactions
### DP-600 Exam Tip
* Warehouse = Read + Write SQL
* Lakehouse SQL Endpoint = Read-Only SQL
### Summary
This is one of the most frequently tested Fabric concepts.

## **14. Database Mirroring in Fabric**
### Explanation
Database Mirroring enables near-real-time replication into OneLake.

Supported Sources:
* Azure SQL Database
* Azure Cosmos DB
* Azure Databricks
* Snowflake
### Benefits
* Near real-time synchronization
* No complex ETL required
* Automatic Delta conversion
* Cross-database querying
### Summary
Database Mirroring provides a managed mechanism to bring external data into Fabric analytics workflows.
