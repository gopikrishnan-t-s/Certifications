## **3. OneLake Features**
### Explanation
OneLake is Fabric's unified storage platform.
### Key Characteristics
#### Single Data Lake
* One unified storage platform.
* Automatically available in every Fabric tenant.
#### Supports Multiple Engines
* Spark
* T-SQL
* Power BI
* Real-Time Analytics
#### Storage Format
* Delta Parquet
* ACID Transactions
* Time Travel
#### Data Types Supported
* Structured
* Semi-Structured
* Unstructured
### OneLake Uses
#### Lakehouses
Store data directly in OneLake.
#### Warehouses
Use OneLake as backend storage.
#### Power BI
Can query data directly through Direct Lake mode.
#### Machine Learning
Acts as centralized training-data storage.
#### Data Pipelines & Dataflows
Used as staging and transformation storage.
### Summary
OneLake is the foundational storage layer for almost every Fabric workload.

## **5. Introducing KQL and KQL DB**
### Explanation
#### KQL (Kusto Query Language)
A read-only query language optimized for:
* Logs
* Telemetry
* Time-series data
* Real-time analytics
#### KQL Database
A dedicated real-time analytics database built around KQL.

Designed for:
* Streaming data
* Operational monitoring
* IoT analytics
* Event-driven systems
### KQL vs SQL
| Feature | KQL | SQL |
| --- | --- | --- |
| Real-Time Analytics | ✅ | Limited |
| Time-Series Analysis | ✅ | Limited |
| Unstructured Data | ✅ | ❌ |
| JSON Support | ✅ | Limited |
| INSERT / UPDATE / DELETE | ❌ | ✅ |
| Transactions | ❌ | ✅ |
| Stored Procedures | ❌ | ✅ |
### Summary
KQL is optimized for querying and analyzing real-time operational data, while SQL is optimized for transactional and relational workloads.

## **7. Ingesting Data into KQL Databases**
### Explanation
#### Streaming Ingestion
Real-time ingestion from:
* Event Hubs
* IoT Hub
* Kafka
* Pub/Sub
##### Characteristics
* Near real-time availability
* Row-store first
* Later moved to columnar storage
#### Batch Ingestion
Sources:
* OneLake
* ADLS
* Azure Blob
* Amazon S3
* Relational Databases
##### Characteristics
* High throughput
* Large-scale loading
* Stored directly in columnar format
### KQL Command-Based Ingestion
Examples:
```kql
.ingest inline

.ingest from query

.ingest from storage
```
### Summary
KQL databases support streaming, batch, manual, and programmatic ingestion methods.

## **8. Transforming, Querying, and Analyzing Data in KQL**
### Explanation
#### Data Transformation Functions
##### `extend`
Add calculated columns.
##### `project`
Select or rename columns.
##### `where`
Filter records.
##### `parse_json`
Process JSON.
##### `summarize`
Aggregate data.
##### `mv-expand`
Expand arrays into rows.
### Querying Data
KQL Queryset supports:
* KQL Queries
* Partial SQL Support
* Saved Queries
* Shared Queries

Basic Example:
```kql
StormEvents
| where State == "FLORIDA"
```
Aggregation Example:
```kql
StormEvents
| summarize count() by State
```
### Time-Series Analytics
Functions:
#### `make-series`
Time series generation.
#### `bin()`
Time bucketing.
#### `ago()`
Relative time filtering.
#### `series_decompose_anomalies`
Anomaly detection.

### Visualization
#### `render`
Built-in chart generation.

Supports:
* Bar Charts
* Line Charts
* Scatter Plots
* Time Charts
#### Power BI Integration
Build dashboards from KQL data.
#### Kusto Explorer
Desktop analytics tool for KQL databases.
### Summary
KQL combines transformation, querying, analysis, anomaly detection, and visualization in a single analytics engine.
