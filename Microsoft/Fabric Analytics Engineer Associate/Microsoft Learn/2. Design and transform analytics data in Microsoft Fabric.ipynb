## **Choose data stores in Microsoft Fabric**

### **Introduction**

#### Choosing Between Lakehouse, Warehouse, and Eventhouse in Microsoft Fabric
Microsoft Fabric provides three primary analytical data stores:
1. Lakehouse
2. Warehouse
3. Eventhouse

All three store data in **OneLake** using open formats, but they are optimized for different workloads, users, and data patterns.
#### At a Glance
| Feature                 | Lakehouse                                 | Warehouse                        | Eventhouse                     |
| ----------------------- | ----------------------------------------- | -------------------------------- | ------------------------------ |
| Primary Data Type       | Structured, semi-structured, unstructured | Structured relational data       | Streaming and time-series data |
| Query Language          | SQL + Spark                               | Full T-SQL                       | KQL (and some T-SQL)           |
| Read/Write SQL          | SQL endpoint is read-only                 | Full read/write                  | Primarily query-focused        |
| Supports Spark          | ✅                                         | ❌                                | ❌                              |
| Real-Time Streaming     | Limited                                   | Limited                          | ✅                              |
| Data Science & ML       | ✅ Excellent                               | Limited                          | Limited                        |
| Traditional BI          | ✅                                         | ✅ Excellent                      | Moderate                       |
| Time-Series Analytics   | Limited                                   | Limited                          | ✅ Excellent                    |
| Fact/Dimension Modeling | Possible                                  | Ideal                            | Not primary use                |
| Best Users              | Data Engineers, Data Scientists           | BI Developers, SQL Professionals | Operations, Monitoring Teams   |

### **Evaluate lakehouse capabilities**

#### What is a Lakehouse?
A **Lakehouse** combines:
* The scalability of a **Data Lake**
* The analytical capabilities of a **Data Warehouse**

It is the most flexible analytical data store in Microsoft Fabric and is built on:
* **OneLake**
* **Delta Lake**
* **Apache Spark**
#### Lakehouse Architecture
A lakehouse contains two primary storage areas:
##### 1. Tables
Stores managed **Delta tables**.

Features:
* Structured and queryable
* Automatically registered in the metastore
* Schema support
* ACID transactions

Benefits:
* No need to manually create tables with CREATE TABLE
* Fabric automatically validates and registers data
##### 2. Files
Stores:
* CSV files
* JSON files
* Parquet files
* Images
* Documents
* Other unstructured content

Features:
* Flexible storage
* Schema not required
* Ideal for raw data
#### Dual Access Model
One of the biggest advantages of a Lakehouse is that the same data can be accessed in two ways.
##### Apache Spark
Used by:
* Data Engineers
* Data Scientists

Languages:
* Python
* PySpark
* Scala
* Spark SQL
* R

Capabilities:
* Ingest data
* Transform data
* Analyze data
* Train machine learning models
##### Read and Write Support
✅ Read \
✅ Write
#### SQL Analytics Endpoint
Used by:
* Analysts
* Report developers
* SQL users

Capabilities:
* Query Delta tables using T-SQL
* Connect Power BI
* Support Direct Lake mode
##### Access Type
✅ Read only \
❌ No `INSERT` \
❌ No `UPDATE` \
❌ No `DELETE`

#### Lakehouse Capabilities Summary
| Capability          | Details                                   |
| ------------------- | ----------------------------------------- |
| Data Types          | Structured, Semi-structured, Unstructured |
| Primary Tool        | Apache Spark                              |
| Read Access         | Spark + SQL                               |
| Write Access        | Spark, Dataflows, Pipelines               |
| Storage Format      | Delta Lake                                |
| Transaction Support | Single-table ACID transactions            |
| SQL Support         | Read-only T-SQL endpoint                  |
| Schema Management   | Schema-on-read with Delta enforcement     |

### **Evaluate warehouse capabilities**

#### What is a Fabric Warehouse?
A **Microsoft Fabric Warehouse** is an enterprise-scale relational analytical database built on OneLake that provides **a full T-SQL experience** with transactional capabilities.

While both Lakehouse and Warehouse store data in Delta format on OneLake, the Warehouse is designed specifically for:
* Structured data
* SQL-first development
* Business Intelligence (BI)
* Dimensional modeling
* Transactional data management
#### Warehouse Architecture
The Warehouse stores data in:
> OneLake → Delta Format → Warehouse Tables

Unlike a Lakehouse, the Warehouse emphasizes **schema-on-write** and **full SQL management**.
#### Core Warehouse Capabilities
##### 1. Full DML Support
The Warehouse supports:
```sql
INSERT
UPDATE
DELETE
MERGE
```
Benefits:

✅ Modify data in place \
✅ Handle Slowly Changing Dimensions (SCD) \
✅ Perform corrective updates \
✅ Support upsert operations using MERGE
##### 2. Full DDL Support
Supports creation and management of database objects:
```sql
CREATE TABLE
ALTER TABLE
DROP TABLE
CREATE VIEW
CREATE PROCEDURE
CREATE FUNCTION
```
Benefits:

✅ Full database development experience \
✅ Familiar SQL Server workflow
##### 3. Multi-Table ACID Transactions
Supports:
> Begin Transaction → Update Multiple Tables → Commit or Rollback

Benefits:

✅ Data consistency \
✅ Referential integrity \
✅ Reliable dimensional modeling

**Important Difference**
| Feature           | Lakehouse    | Warehouse   |
| ----------------- | ------------ | ----------- |
| ACID Transactions | Single-table | Multi-table |

##### 4. Cross-Database Queries
Can query:
* Other Warehouses
* Lakehouse SQL Endpoints

Using:
```
database.schema.table
```
Example:
```sql
SELECT *
FROM SalesWarehouse.dbo.FactSales;
```
Benefits:

✅ No data copying \
✅ Unified analytics \
✅ Cross-workload access

### **Evaluate eventhouse capabilities**

#### What is an Eventhouse?
An **Eventhouse** is Microsoft Fabric's analytical data store for:
* Real-time analytics
* Streaming data
* Time-series analysis
* Telemetry and monitoring workloads

It's optimized for continuously arriving events and provides very fast analytics on large volumes of time-based data.
##### Best Fit
✅ IoT Sensors \
✅ Application Logs \
✅ Telemetry \
✅ Website Events \
✅ Operational Monitoring \
✅ Real-Time Dashboards
#### Eventhouse Architecture
An Eventhouse contains one or more KQL Databases.
> Event Sources → Eventstream / Connectors → Eventhouse → KQL Database → KQL Queries → Dashboards / Alerts

##### Key Design Characteristic
Data is automatically:
* Indexed
* Compressed
* Partitioned by ingestion time

This allows rapid querying even against massive event volumes.
#### Core Eventhouse Capabilities
| Capability       | Details                                          |
| ---------------- | ------------------------------------------------ |
| Data Types       | Time-series, Events, Semi-structured, Structured |
| Primary Language | KQL (Kusto Query Language)                       |
| Write Pattern    | Streaming & Batch Ingestion                      |
| Read Pattern     | KQL, limited T-SQL, Power BI                     |
| Storage          | Columnar storage                                 |
| Indexing         | Automatic                                        |
| Transactions     | Append-optimized                                 |
| Schema           | Schema-on-write with dynamic fields              |

#### Query Language: KQL
The primary language is:
##### Kusto Query Language (KQL)
Designed specifically for:
* Time-series analytics
* Log analytics
* Monitoring data
* Event processing

Example:
```kql
SensorData
| where Timestamp > ago(1h)
| summarize AvgTemp = avg(Temperature)
    by DeviceId
```
Advantages:

✅ Fast aggregations \
✅ Time-window analysis \
✅ Pattern matching \
✅ Anomaly detection \
✅ Geospatial analysis

#### Data Ingestion
An Eventhouse can ingest data from:
##### Eventstreams
Most common Fabric approach.
##### SDKs
Application-generated events.
##### Apache Kafka
Streaming platforms.
##### Logstash
Log ingestion scenarios.
##### Dataflows
Batch-to-stream integration.

#### When to Choose an Eventhouse
##### 1. IoT and Sensor Data
Examples:
* Temperature sensors
* Manufacturing equipment
* Distribution center monitoring
* Smart devices

Why?

✅ Continuous event ingestion \
✅ Time-series optimization \

Dimension tables are split into related tables.
##### Example

```
DimProduct
```

```
DimProduct
      |
----------------
|              |
Category    Supplier
```
##### Structure
```
FactSales
    |
DimProduct
    |
Category
```

When dimension tables become extremely large.

Benefits:

**More Complexity**

Harder for:

### **Design fact tables**

#### What is a Fact Table?
A **Fact Table** is the central table in a dimensional model that stores:
* Business events
* Measurements
* Metrics used for analysis

Examples:
* Sales transactions
* Orders
* Inventory counts
* Deliveries
* Sensor readings

Think:

✅ Facts = Numbers \
✅ Facts = Events
#### Fact Table Structure
A well-designed fact table contains **three types of columns**:
##### 1. Dimension Keys
Foreign keys that connect to dimension tables.

Examples:
```
CustomerKey
ProductKey
DateKey
StoreKey
```
Purpose:
* Provide context
* Enable filtering and grouping

Answer business questions such as:
> Who? \
> What? \
> When? \
> Where?

##### 2. Measures
Numeric values that can be analyzed.

Examples:
```
SalesAmount
Quantity
Cost
Profit
Duration
```
These are the values users aggregate.

Example:
```sql
SUM(SalesAmount)
AVG(Cost)
COUNT(Orders)
```
##### 3. Attributes (Degenerate Dimensions)
Additional descriptive values about the event.

Examples:
```
OrderNumber
TrackingNumber
InvoiceNumber
```
These describe the event but usually don't need separate dimension tables.

They are called:

**Degenerate Dimensions**

Because they behave like dimensions but stay inside the fact table.

#### Fact Table Design Rule
A good fact table should be:

✅ Narrow \
✅ Mostly keys and measures \
✅ Free from descriptive text columns

Descriptive data belongs in dimensions.

#### The Most Important Concept: Grain
##### What is Grain?
Grain defines exactly what one row represents.
##### Examples
**Sales Fact**

One row per sales order line

**Inventory Fact**

One row per product per day

**Sensor Fact**

One row per hourly sensor reading
##### Why Grain Matters
The grain determines:
* Which dimensions belong in the table
* Which measures belong in the table
* How data can be analyzed
##### Exam Rule
✅ Always define the grain first.

Before:
* Dimensions
* Measures
* Keys
##### Grain Too High
Example:
> One row per day of sales

**Problem**:

Cannot later analyze:
* Individual products
* Individual customers
* Individual orders

Loss of detail.
##### Grain Too Low
Example:
> One row every second

for a sensor that changes only hourly.

Problem:
* Excess storage
* Poor efficiency

Too much unnecessary detail.

#### Types of Fact Tables
##### 1. Transaction Fact Table
**Purpose**

Stores one row per business event.

**Examples**:
* Sales transactions
* Website clicks
* Support tickets

**Characteristics**

✅ Lowest level of detail \
✅ Append-only \
✅ Event-based

**Example**
> Order Line 1 \
> Order Line 2 \
> Order Line 3

Each line becomes a separate fact row.

##### 2. Periodic Snapshot Fact Table
**Purpose**

Captures status at regular intervals.

**Examples**:
* Daily inventory
* Monthly account balance
* Weekly staffing levels

**Characteristics**

✅ One row per time period \
✅ Supports trend analysis \
✅ Focused on state rather than events

**Example**
> Inventory on Jan 1 \
> Inventory on Jan 2 \
> Inventory on Jan 3

##### 3. Accumulating Snapshot Fact Table
**Purpose**

Tracks a process through milestones.

**Examples**:
* Order fulfillment
* Loan processing
* Claims management

**Characteristics**

✅ Row created once \
✅ Updated as process progresses \
✅ Multiple milestone dates

**Example**
> Order Date \
> Approval Date \
> Ship Date \
> Delivery Date

all in the same row.
#### Measure Types
Measures are classified based on aggregation behavior.
##### Additive Measures
Can be summed across all dimensions.

Examples:
> Revenue \
> Sales Amount \
> Quantity \
> Cost

Example:
```sql
SUM(Revenue)
```
works everywhere.
##### Semi-Additive Measures
Can be summed across some dimensions but not time.

Examples:
> Inventory Level \
> Account Balance

Problem:
```
Inventory Jan 1 = 100
Inventory Jan 2 = 120

100 + 120 = 220 ❌
```
Doesn't make business sense across time.
##### Non-Additive Measures
Cannot be summed meaningfully.

Examples:
```
Temperature
Discount %
Unit Price
Ratios
```
Instead of storing:
```
Discount %
```
store:
> Discount Amount \
> Sales Amount

and calculate the percentage later.
#### Naming Conventions
Common naming patterns:
```
Fact_Sales
Fact_Inventory
Fact_Order
```
or
```
f_Sales
f_Inventory
```
Benefits:

✅ Easy identification \
✅ Consistent design \
✅ Better discoverability

#### Factless Fact Tables
Sometimes events have no numeric measures.

Examples:
> Student attended class \
> Customer visited website \
> Promotion applied

No sales amount or quantity exists.
##### Solution
Use a **Factless Fact Table**.

Analysis happens by counting rows.

Example:
```sql
COUNT(*)
```
#### Aggregate Fact Tables
Aggregate facts store precomputed summaries.
##### Example:
Instead of:
> Daily Sales Transactions

store:
> Monthly Sales Totals

Benefits:

✅ Faster reports \
✅ Better performance \
✅ Reduced query complexity
##### Example
> FactSales (Daily) → AggregateSales (Monthly)

### **Design dimension tables**

#### What is a Dimension Table?
A **Dimension Table** provides the context for the measurements stored in fact tables.

Dimensions answer:
* **Who?** → Customer
* **What?** → Product
* **When?** → Date
* **Where?** → Geography/Store
* **Why?** → Business classifications

Without dimensions:
```
Revenue = 100,000
```
With dimensions:
```
Revenue = 100,000
by Product
by Region
by Quarter
```
Dimensions make analytics meaningful.
#### Structure of a Dimension Table
A dimension table typically contains:
##### 1. Surrogate Key
A warehouse-generated unique identifier.

Example:
```
CustomerKey = 1001
```
Characteristics:

✅ Primary Key \
✅ Used in fact table relationships \
✅ Independent of source systems
##### 2. Natural Key (Business Key)
The identifier that comes from the source system.

Example:
```
CustomerID = CUST-001
ProductCode = PROD-100
```
Purpose:

✅ Links warehouse data back to source systems \
✅ Used during ETL processes
##### 3. Dimension Attributes
Descriptive fields used for filtering and grouping.

Example:
```
CustomerName
Region
Customer Segment
Loyalty Tier
```
These are the columns report users typically interact with.
#### Why Use Surrogate Keys?
Even when a natural key exists, surrogate keys are preferred.
##### Benefits
1. Avoid Source-System Conflicts

Two source systems may use the same business key.

Example:
```
System A → Customer 100
System B → Customer 100
```
Surrogate keys eliminate ambiguity.

2. Replace Complex Keys

Instead of:
> Country + CustomerID + Division

you use:
```
CustomerKey
```
Simpler and faster.

3. Support Slowly Changing Dimensions

Multiple versions of the same customer can exist.

Example:
```
CustomerKey = 1001 (Old Region)
CustomerKey = 1002 (New Region)
```
4. Reduce Fact Table Size

Small integer keys consume less storage.

Example:
```
1001
```
instead of:
```
CUSTOMER-ABC-000123
```
#### Date Dimension Exception
The Date dimension commonly uses:
```
YYYYMMDD
```
Example:
```
20260115
```
This acts as both:

✅ Surrogate Key \
✅ Meaningful Business Value
#### Common Dimension Tables
##### Date Dimension
Supports time analysis.

Typical Attributes:
* Year

##### Customer Dimension
Supports customer analysis.

#### How to Identify Needed Dimensions
Listen for the word:

**"BY"**

Example:
> Sales by Region \
> Sales by Product \
> Sales by Month

This means you need:
* Region Dimension
* Product Dimension

#### Denormalize Dimensions
Dimension tables are usually denormalized.
##### Recommended
```
DimProduct
  Category
  Subcategory
  Brand
```
##### Not Recommended
```
DimProduct
     ↓
DimCategory
```
##### Reason:
✅ Fewer joins \
✅ Faster reporting \
✅ Simpler Power BI models

#### Hierarchies
Hierarchies enable drill-down analysis.
##### Balanced Hierarchy
Every branch has the same levels.

Example:
> Year → Quarter → Month → Day

Most common hierarchy type.
##### Unbalanced Hierarchy
Branches have different depths.

Example:
> CEO → VP → Manager

Some branches may skip levels.
##### Ragged Hierarchy
Some levels don't exist for all members.

Example:
> Country → State → City

Certain countries may not have states.
#### Conformed Dimensions
A **Conformed Dimension** is shared by multiple fact tables.
##### Example:
```
FactSales → DimProduct
```
```
FactInventory → DimProduct
```
##### Benefits:
✅ Consistency \
✅ Easier cross-process analysis \
✅ Single source of truth
##### Example
Analyze:
> Sales by Product \
> Inventory by Product

using one shared Product dimension.
#### Role-Playing Dimensions
A **Role-Playing Dimension** is reused multiple times within one fact table.
##### Example:
One Date Dimension:
```
DimDate
```
Referenced as:
> Order Date \
> Ship Date \
> Delivery Date

inside the same fact table.
##### Benefits:
✅ Avoids duplicate dimension tables \
✅ Simplifies maintenance
#### Junk Dimensions
Used when there are many small attributes with few values.

Examples:
> Order Status \
> Promotion Flag \
> Online Order Flag \
> Member Flag

Instead of creating many tiny dimensions:
```
DimStatus
DimPromotion
DimFlag
```
combine them into:
```
DimJunk
```
##### Benefits:
✅ Fewer foreign keys \
✅ Simpler model \
✅ Cleaner fact tables
#### Naming Conventions
Common naming patterns:
```
Dim_Customer
Dim_Product
Dim_Date
```
or
```
d_Customer
d_Product
```
##### Benefits:
✅ Easy identification \
✅ Consistent model design

✅ Use surrogate keys \

✅ Denormalize whenever possible \
✅ Build useful hierarchies \
✅ Share dimensions across facts (Conformed Dimensions) \

### **Implement slowly changing dimensions**

#### What are Slowly Changing Dimensions (SCDs)?
Dimension data changes over time.

Examples:
* Customer changes address
* Employee changes department
* Product changes category
* Salesperson changes region

A **Slowly Changing Dimension (SCD)** defines how a dimensional model handles these changes while balancing historical accuracy, storage, and complexity.
#### Why SCDs Matter
Suppose a salesperson moves from:
> East Region → West Region

If historical sales are analyzed later:
##### Question:
Should old sales stay attributed to East Region (historically correct)? \
OR \
Should all sales move to West Region (current view)?

**The answer determines which SCD type to use.**
#### SCD Type 0 (Retain Original)
##### Description
* Never allow changes.
* The original value is permanently preserved.
##### Use Cases
* Date of Birth
* Original Credit Score
* Initial Registration Date
##### Example
```
CustomerDOB = 01-Jan-1990
```
Even if source systems attempt a change:
> Value remains unchanged.

##### Characteristics
✅ No updates \
✅ No history tracking needed \
✅ Fixed reference data

#### SCD Type 1 (Overwrite)
##### Description
* Replace old values with new values.
* No history is retained.
##### Example
Before:
```
CustomerID = 100
Email = old@email.com
```
After:
```
CustomerID = 100
Email = new@email.com
```
Old value disappears.
##### When to Use
✅ Data corrections \
✅ Historical values aren't important \
✅ Simplest maintenance
##### Drawback
* Historical reporting changes.

Example:
> Salesperson Region: East → West

After overwrite:
> All historical sales appear in West.

Potentially misleading.
#### SCD Type 2 (Most Important for Exams)
##### Description
* Insert a new row whenever a tracked attribute changes.
* The original row remains.
* Every version receives its own surrogate key.
##### Example
Before
| CustomerKey | CustomerID | Region |
| ----------- | ---------- | ------ |
| 1001        | C001       | East   |

After move to West
| CustomerKey | CustomerID | Region |
| ----------- | ---------- | ------ |
| 1001        | C001       | East   |
| 1002        | C001       | West   |

Both versions exist.
#### Additional Columns Required
##### Start Date
When this version became active.
##### End Date
When the version expired.

Current row often uses:
```
9999-12-31
```
##### Current Flag
Identifies active record.
```
TRUE
FALSE
```
#### Type 2 ETL Process
##### Step 1
Expire existing row.
```
Set EndDate
Set CurrentFlag = FALSE
```
##### Step 2
Insert new row.
```
New Surrogate Key
New Attributes
CurrentFlag = TRUE
```
##### Benefits
✅ Full history \
✅ Accurate historical reporting \
✅ Most common warehouse approach \
✅ Supports audit requirements
##### Example Use Cases
* Customer region changes
* Employee department changes
* Product category changes
#### SCD Type 3 (Add Column)
##### Description
Store current and previous value.
##### Example
| Employee | CurrentRegion | PreviousRegion |
| -------- | ------------- | -------------- |
| John     | West          | East           |

##### Characteristics
✅ Tracks one previous change \
✅ Simple implementation \
❌ Loses older history
##### Use Cases
When you only need:
> Current vs Previous

comparison.
#### SCD Type 6 (Hybrid)
##### Description
Combines:
* Type 1
* Type 2
* Type 3
##### Provides
✅ Full historical versions \
✅ Current value available everywhere \
✅ Previous value tracking
##### Benefit
Can easily answer:
* Current Region?
* Historical Region?
* Previous Region?

from the same model.
##### Drawback
❌ Complex ETL \
❌ More maintenance \
❌ More storage
#### Choosing the Right SCD Type
| Requirement                       | Recommended Type |
| --------------------------------- | ---------------- |
| Never changes                     | Type 0           |
| Correction only                   | Type 1           |
| Full history required             | Type 2           |
| Current vs Previous               | Type 3           |
| Current + Previous + Full History | Type 6           |

#### SCD Example Comparison
##### Customer Moves Regions
**Type 1**
> East → West

Result:
> Only West remains.

**Type 2**
> East Version Retained \
> West Version Created

Result:
> Full historical accuracy.

**Type 3**
> Current Region = West \
> Previous Region = East

Result:
> One previous change preserved.

#### Tradeoffs
##### Storage
**Type 1**

✅ Small

**Type 2**

❌ Grows continuously

because new rows are added.
##### Query Complexity
**Type 1**

✅ Easy

**Type 2**

❌ Requires:
* Effective Date checks
* Current Flags
* Historical joins

##### ETL Complexity
**Type 1**

✅ Simple update

**Type 2**

❌ Detect changes ❌ Expire records ❌ Insert new versions

##### Business Needs
Most important rule:
* Don't track history unless the business needs it.

And:
* Don't ignore history when the business requires historical analysis.

#### Best Practice
Use:
##### Type 2
Only for attributes where history matters.

Example:

Track:

✅ Customer Region \
✅ Employee Department \
✅ Product Category

Do NOT track:

❌ Formatting changes \
❌ Minor corrections

Use Type 1 for those.

This keeps dimensions manageable.

#### Exam Quick Facts
✅ SCD manages changes in dimension data over time \
✅ Type 0 = Never change \
✅ Type 1 = Overwrite existing value \
✅ Type 2 = Add new row (full history) \
✅ Type 3 = Add new column (previous value) \
✅ Type 6 = Hybrid approach \
✅ Type 2 requires:
* Start Date
* End Date
* Current Flag
* New Surrogate Keys

✅ Type 2 is the most common SCD implementation in data warehouses \
✅ Warehouse surrogate keys make Type 2 possible

#### Memory Trick
* 0 = Freeze
* 1 = Replace
* 2 = New Row ⭐ (Most Important)
* 3 = New Column
* 6 = Everything

For exams, if the requirement says:
> "Maintain full historical accuracy"

The answer is almost always:

**✅ SCD Type 2**.

The fact table captures the business events you want to measure — for Contoso Retail, that’s sales transactions. The grain is **one row per sales transaction line item**. The numeric columns you aggregate in queries are called **measures**: 
* quantity 
* unit price
* sales amount 
* discount amount.

1. In your warehouse, select the **New SQL query** button on the toolbar, and enter the following T-SQL statement:

```sql
CREATE TABLE f_Sales
(
    DateKey INT NOT NULL,
    StoreKey INT NOT NULL,
    ProductKey INT NOT NULL,
    CustomerKey INT NOT NULL,
    Quantity INT NOT NULL,
    UnitPrice DECIMAL(10,2) NOT NULL,
    SalesAmount DECIMAL(10,2) NOT NULL,
    DiscountAmount DECIMAL(10,2) NOT NULL
);
```

#### Create the dimension tables
Dimension tables provide the context that makes fact data meaningful. They answer the who, what, when, and where behind every measurement. For this model, you need four dimensions: date, store, product, and customer.

The product and store dimensions include SCD Type 2 tracking columns (`ValidFrom`, `ValidTo`, `IsCurrent`) because the business needs to track historical changes to product costs and store regional assignments. The customer and date dimensions use simpler structures because only corrections (Type 1) are needed for customers, and the date dimension is static reference data.
1. On the **Home** menu tab, select **New SQL query** and run the following code to create all four dimension tables:

```sql
-- Date dimension: uses YYYYMMDD integer format as surrogate key
CREATE TABLE d_Date
(
    DateKey INT NOT NULL,
    FullDate DATE NOT NULL,
    [Year] INT NOT NULL,
    [Quarter] INT NOT NULL,
    [Month] INT NOT NULL,
    MonthName VARCHAR(10) NOT NULL,
    [Day] INT NOT NULL,
    [DayOfWeek] VARCHAR(10) NOT NULL,
    FiscalYear INT NOT NULL,
    FiscalQuarter INT NOT NULL,
    IsHoliday BIT NOT NULL,
    IsWeekday BIT NOT NULL
);
```
```sql
-- Store dimension: includes SCD Type 2 tracking columns
CREATE TABLE d_Store
(
    StoreKey INT NOT NULL,
    StoreNaturalKey VARCHAR(10) NOT NULL,
    StoreName VARCHAR(50) NOT NULL,
    StoreType VARCHAR(20) NOT NULL,
    City VARCHAR(50) NOT NULL,
    [State] VARCHAR(50) NOT NULL,
    Country VARCHAR(50) NOT NULL,
    Region VARCHAR(50) NOT NULL,
    OpenDate DATE NOT NULL,
    ValidFrom DATE NOT NULL,
    ValidTo DATE NOT NULL,
    IsCurrent BIT NOT NULL
);
```
```sql
-- Product dimension: includes SCD Type 2 tracking columns
CREATE TABLE d_Product
(
    ProductKey INT NOT NULL,
    ProductNaturalKey VARCHAR(10) NOT NULL,
    ProductName VARCHAR(50) NOT NULL,
    Brand VARCHAR(50) NOT NULL,
    Subcategory VARCHAR(50) NOT NULL,
    Category VARCHAR(50) NOT NULL,
    UnitCost DECIMAL(10,2) NOT NULL,
    ValidFrom DATE NOT NULL,
    ValidTo DATE NOT NULL,
    IsCurrent BIT NOT NULL
);
```

#### Add table constraints
Now that the fact and dimension tables exist, connect them into a star schema by adding foreign key constraints, loading sample data, and running analytical queries. In Fabric Warehouse, table constraints (primary keys and foreign keys) can’t be defined inline within a `CREATE TABLE` statement — use `ALTER TABLE` to add them after the tables are created. Constraints are `NOT ENFORCED` and serve as metadata that documents the relationships between tables. This metadata helps Power BI auto-detect relationships when you create a semantic model from the warehouse.
1. Create a new SQL query and run the following code to add primary keys to each dimension table and foreign keys to the fact table:

```sql
-- Add primary keys to dimension tables
ALTER TABLE d_Date
    ADD CONSTRAINT PK_d_Date PRIMARY KEY NONCLUSTERED (DateKey) NOT ENFORCED;

ALTER TABLE d_Store
    ADD CONSTRAINT PK_d_Store PRIMARY KEY NONCLUSTERED (StoreKey) NOT ENFORCED;

ALTER TABLE d_Product
    ADD CONSTRAINT PK_d_Product PRIMARY KEY NONCLUSTERED (ProductKey) NOT ENFORCED;

ALTER TABLE d_Customer
    ADD CONSTRAINT PK_d_Customer PRIMARY KEY NONCLUSTERED (CustomerKey) NOT ENFORCED;

-- Add foreign keys to the fact table
ALTER TABLE f_Sales
    ADD CONSTRAINT FK_Sales_Date FOREIGN KEY (DateKey)
        REFERENCES d_Date(DateKey) NOT ENFORCED;

ALTER TABLE f_Sales
    ADD CONSTRAINT FK_Sales_Store FOREIGN KEY (StoreKey)
        REFERENCES d_Store(StoreKey) NOT ENFORCED;

ALTER TABLE f_Sales
    ADD CONSTRAINT FK_Sales_Product FOREIGN KEY (ProductKey)
        REFERENCES d_Product(ProductKey) NOT ENFORCED;

ALTER TABLE f_Sales
    ADD CONSTRAINT FK_Sales_Customer FOREIGN KEY (CustomerKey)
        REFERENCES d_Customer(CustomerKey) NOT ENFORCED;
```

#### Implement SCD patterns
Dimension data changes over time. Customers move, products get repriced, and stores get reassigned to different regions. Slowly changing dimension (SCD) patterns define how your dimensional model responds to these changes.

The product dimension uses two SCD patterns:
* **Type 2 (add new row)** for `UnitCost` — the business needs to track cost changes for historical margin analysis.
* **Type 1 (overwrite)** for `ProductName` — name corrections should apply to all history.
##### Simulate an SCD Type 2 change
Suppose the cost of the Mountain Bike Pro increases from $1,200 to $1,350 effective March 1, 2026. An SCD Type 2 change expires the current row and inserts a new version.
1. Create a new SQL query and run the following code to:
    * Expire the current product version
    * Insert the new version
    * Add a sale that references the updated product
```sql
-- Step 1: Expire the current version of Mountain Bike Pro
UPDATE d_Product
SET ValidTo = '2026-03-01',
    IsCurrent = 0
WHERE ProductNaturalKey = 'MB-PRO'
  AND IsCurrent = 1;

-- Step 2: Insert the new version with updated cost
INSERT INTO d_Product VALUES
(6, 'MB-PRO', 'Mountain Bike Pro', 'AdventureWorks', 'Mountain Bikes', 'Bikes', 1350.00, '2026-03-01', '9999-12-31', 1);

-- Step 3: A sale after the cost change references the new product version (ProductKey = 6)
INSERT INTO f_Sales VALUES
(20260504, 1, 6, 5, 1, 1500.00, 1500.00, 0.00);
```
2. Create a new SQL query and run the following code to see how SCD Type 2 preserves historical accuracy:
```sql
SELECT
    d.FullDate,
    p.ProductName,
    p.UnitCost AS ProductCostVersion,
    p.ValidFrom AS CostEffectiveDate,
    f.Quantity,
    f.SalesAmount
FROM f_Sales f
JOIN d_Date d ON f.DateKey = d.DateKey
JOIN d_Product p ON f.ProductKey = p.ProductKey
WHERE p.ProductNaturalKey = 'MB-PRO'
ORDER BY d.FullDate;
```
> Review the results. The January, February, and April sales are linked to the original cost version ($1,200), while the May sale is linked to the new cost version ($1,350). Each fact row retains the product cost that was in effect at the time of the sale. This is the key benefit of SCD Type 2 — historical facts remain accurate even after dimension attributes change.

#### 2. Query Folding
##### What Is Query Folding?
Query Folding occurs when Power Query pushes transformations back to the source system for execution.

Instead of:
> Source Data → Download Everything → Power Query Processes All Data

Power Query sends transformation logic to the source:
> Power Query → SQL Query Generated → Source Executes Query → Only Results Returned

#### Why Query Folding Is Faster
##### Source Systems Are Optimized
Databases are built to:
* Filter
* Aggregate
* Sort

more efficiently than transferring large datasets.
##### Less Network Traffic
Instead of:
```
1,000,000 rows transferred
```
only:
```
Filtered rows transferred
```
##### Less Power Query Processing
Reduces memory and compute requirements.
#### Query Folding Outcomes
##### Full Folding
> All transformations → Executed by source

Best outcome.

✅ Highest performance
##### Partial Folding
> Some transformations → Source

> Remaining transformations → Power Query

Acceptable.
##### No Folding
> All raw data downloaded → Everything processed locally

Worst outcome.

❌ Slowest performance \
❌ Highest resource usage

#### Apply Power Query transformations
These transformations represent the most common data quality and preparation tasks that data professionals perform to convert raw source data into analytics-ready tables. In this section, you apply column selection, filtering, data type changes, renaming, and a calculated column to the orders data.

The first three tasks support query folding, where Power Query pushes operations back to the data source for more efficient execution. Placing these steps first maximizes the work that can be folded. Steps that follow, like renaming and custom columns, typically break the fold boundary and run locally in the Power Query engine.
> Tip: In the Query Settings pane on the right side, notice that each transformation appears as a step in Applied Steps. You can select any step to see the data at that point in the transformation process.

##### Choose columns
Removing extra columns reduces the data retrieved from the source and is one of the most effective steps for query folding. In this task, you keep only the columns needed for analysis.

    * `CustomerID`

    * `OrderQty`

##### Filter rows
Filtering rows before other transformations reduces the volume of data that subsequent steps process and folds well to most data sources. In this task, you remove rows with missing dates and rows with zero or negative quantities.

2. Select the `OrderQty` column header. Select the drop-down arrow, select **Number Filters**, and then select **Greater Than**. Enter `0` as the value and select **OK**.

Correct data types are essential for calculations, sorting, and filtering, and they enable Copilot and other AI features to interpret your data correctly. Data type changes also fold to many sources, so applying them before rename and text operations keeps the fold chain intact. In this task, you verify the data types for three columns.
1. On the ribbon, select the **Transform** tab. Select the following columns and make sure the data type is correctly set:

#### What are Fabric Notebooks?
A **Fabric Notebook** is a web-based, interactive coding environment powered by Apache Spark.

A notebook consists of **cells**.

Each cell can contain:
* Code
* Markdown documentation

Example:
```py
df = spark.read.table("Sales")
df.show()
```
When a cell runs:
* Results appear immediately below the cell.

#### Spark Session
Every notebook runs inside a Spark Session.
##### Key Points
✅ Starts automatically when the first code cell runs \
✅ Provides distributed computing \
✅ Shared across all notebook cells \
✅ Persists until stopped or timed out

Example:
```py
customer_count = 100
```
A later cell can still use:
```py
print(customer_count)
```
because all cells share the same Spark session.
