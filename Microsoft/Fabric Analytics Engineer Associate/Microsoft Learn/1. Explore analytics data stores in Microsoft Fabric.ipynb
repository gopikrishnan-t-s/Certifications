## **Introduction to end-to-end analytics using Microsoft Fabric**

### **Introduction**

**Microsoft Fabric** is an end-to-end analytics platform that unifies the entire data lifecycle in a single environment. Instead of using separate tools for data ingestion, preparation, governance, analytics, and AI, organizations can perform all these tasks within Fabric.
#### Why Fabric Matters
* Reduces complexity caused by disconnected data tools and teams.
* Provides a single platform for data engineers, analysts, data scientists, and business users to collaborate.
* Eliminates duplicated effort by using the same governed data across analytics and AI workloads.
#### Core Foundation: OneLake
Fabric is built on **OneLake**, a unified data lake that serves as the central storage layer for organizational data. Because data is stored and managed in one place, it can be used consistently across all Fabric experiences.
#### Fabric Workloads
Fabric brings together capabilities for:
* Data ingestion
* Data engineering
* Data integration
* Data warehousing
* Data science
* Real-time analytics
* Business intelligence and reporting
#### Analytics and AI Together
A key benefit of Fabric is that the data prepared and governed for analytics is automatically available for AI scenarios, including:
* Copilot
* Data agents
* Fabric IQ
* Machine learning models
* Intelligent agents

This means governance, quality, and preparation work done for reporting and dashboards also supports AI initiatives, creating a single trusted data foundation.
#### In One Sentence
Microsoft Fabric is a unified analytics and AI platform built on OneLake that enables organizations to ingest, manage, govern, analyze, and use data for AI workloads from a single integrated environment.

#### 2. OneLake: The Foundation of Fabric
**OneLake** is Fabric's unified, centralized data lake.
##### Key characteristics:
* Built on **Azure Data Lake Storage Gen2 (ADLS Gen2)**.
* Stores data once and makes it available across all Fabric workloads.

### **Understand OneLake**

**OneLake** is the foundational storage layer of Microsoft Fabric.
##### Key characteristics:
* **Tenant-wide data lake** automatically available when Fabric is enabled.
* Provides **a single copy** of data for the entire organization.
* Shared by all Fabric workloads.
* Eliminates data silos and unnecessary data duplication.
* Acts as a single source of truth.
##### Why It Matters
In traditional systems, different teams often maintain separate copies of the same data. With OneLake:
* All workloads access the same underlying data.
* Updates are immediately available to everyone.
* Storage costs and data inconsistencies are reduced.
#### OneLake Catalog
The **OneLake Catalog** is a searchable inventory of data assets across the Fabric tenant. You can:
* Search by asset name.
* Browse by workspace.
* Browse by domain.
* View metadata.
* View asset owners.
* Track lineage.
##### Governance Features
OneLake integrates with Microsoft Purview to provide:
* Data classification
* Sensitivity labels
* Data lineage tracking
* Access control and security

##### 1. Tables
Stored in:
* Lakehouses
* Warehouses
* Eventhouses
##### 2. Files
Supported formats include:
* Parquet
* CSV

##### 3. Shortcuts
References to data stored elsewhere, such as:
* Azure Data Lake Storage

##### 4. Semantic Models
Used by:
* Power BI

OneLake uses **open formats**, meaning data is not locked into proprietary storage formats.

* Parquet files

Benefits:
* Interoperability

### **Discover streaming data in Real-Time hub**

#### Why Real-Time Hub?
Unlike batch data stored in lakehouses and warehouses, streaming data arrives continuously:
##### Examples:
* Online purchases
* IoT sensor readings
* Application logs
* Inventory updates
* Clickstream events

The **Real-Time Hub** helps you discover and manage these streaming data sources across Fabric.

#### 1. What is Real-Time Hub?
**Real-Time Hub** is the centralized catalog for streaming data in Microsoft Fabric.
##### Comparison
| OneLake Catalog | Real-Time Hub |
| --- | --- |
| Batch data | Streaming data |
| Lakehouses | Eventstreams |
| Warehouses | KQL Tables |
| Semantic Models | Eventhouses |
##### Remember:
* OneLake Catalog = Stored data
* Real-Time Hub = Streaming data
#### 2. Key Components
##### Eventstreams
An **eventstream** is a continuous flow of real-time events.

Common sources:
* Azure Event Hubs
* Azure IoT Hub
* Apache Kafka
* Database CDC
* Custom applications

Examples:
* Customer purchases
* Sensor readings
* Web clicks
##### Eventhouses
An **eventhouse** is a container that stores streaming data.

It contains:
* One or more KQL databases
* Time-series and event-based data

Purpose:
* Organize and manage streaming analytics data.
##### KQL Databases and KQL Tables
KQL databases:
* Store streaming data.

KQL tables:
* Contain ingested events.

Benefits:
* Automatically indexed
* Optimized for high-speed analytics
* Support near real-time querying
##### Kusto Query Language (KQL)
KQL is used to:
* Query streaming data
* Analyze events
* Create real-time dashboards
* Generate alerts
#### 3. Discover Streaming Data

#### Query data using the SQL analytics endpoint
Every lakehouse includes a SQL analytics endpoint that provides read-only T-SQL access to tables. This endpoint enables you to query lakehouse data using familiar SQL syntax, making it easy to explore data and verify it contains what you need.

1. At the top-right of your **analytics** lakehouse page, locate the drop-down menu that currently shows **Lakehouse**. Select the drop-down and switch to **SQL analytics endpoint**.



The view changes to a SQL-focused interface where you can query tables using T-SQL.

2. In the toolbar, select **New SQL query** to open a query editor.
3. In the query editor, enter the following T-SQL query:

    ```sql
    SELECT 
         Item,
         SUM(Quantity * UnitPrice) AS TotalRevenue,
         SUM(Quantity) AS TotalQuantity
    FROM sales
    GROUP BY Item
    ORDER BY TotalRevenue DESC;
    ```
This query calculates total revenue and total quantity sold for each item.

4. Select **Run** to execute the query. Review the results showing revenue by item.



5. Write a second query to find the top customers by revenue:

    ```sql
    SELECT TOP 5
        CustomerName,
        SUM(Quantity) AS TotalQuantity,
        SUM(Quantity * UnitPrice) AS TotalRevenue
    FROM sales
    GROUP BY CustomerName
    ORDER BY TotalRevenue DESC;
    ```
Run the query and review the results.

## **Get started with lakehouses in Microsoft Fabric**

### **Introduction**

#### What is a Lakehouse?
A **Lakehouse** is a storage and analytics architecture that combines:
* **Data Lake capabilities** → store large volumes of structured, semi-structured, and unstructured data.
* **Data Warehouse capabilities** → query, analyze, and report on data efficiently.

In Microsoft Fabric, lakehouses are built on OneLake and use:
* **Apache Spark** for data engineering and large-scale processing.
* **SQL compute engines** for querying and analytics.
#### Why Organizations Need Lakehouses
Traditional data warehouses work well for structured data such as:
* Sales transactions
* Customer records
* Inventory data

However, modern organizations also generate:
* Application logs
* IoT sensor data
* Social media data
* JSON and CSV files
* External data feeds

Managing these different data types often requires separate systems, creating:
* Data silos
* Complex integrations
* Increased maintenance costs

A lakehouse solves this by providing a single platform for all data types.
#### Lakehouse Benefits
✅ Store structured, semi-structured, and unstructured data \
✅ Built on the centralized **OneLake** storage layer \
✅ Supports both **SQL** and **Spark** workloads \
✅ Reduces data movement and duplication \
✅ Scales to large data volumes \
✅ Supports analytics, reporting, machine learning, and AI

### **Describe lakehouse features and capabilities**

#### Why Lakehouses?
Traditional architectures force a choice:
| Data Lake                        | Data Warehouse                   |
| -------------------------------- | -------------------------------- |
| Flexible storage                 | Strong analytics                 |
| Handles any data type            | Structured data only             |
| Scalable                         | Optimized SQL querying           |
| Limited governance and analytics | Higher cost and less flexibility |

A **Lakehouse** combines the benefits of both.
##### Lakehouse = Data Lake + Data Warehouse
Benefits:
* Store all data types.
* Query with SQL.
* Process with Spark.
* Support analytics, BI, machine learning, and AI.
#### Lakehouse Structure
A Fabric Lakehouse contains two main areas:
##### 1. Tables Folder
Contains **Delta Lake tables**.

Features:
* Supports SQL queries via SQL Analytics Endpoint.
* Schema enforcement.
* ACID transactions.
* Accessible in Power BI.
* Automatic optimization and maintenance.

Use for:
* Reporting
* Analytics
* Semantic models
* BI workloads
##### 2. Files Folder
Contains raw files in their native format.

Supported formats:
* CSV
* JSON
* Parquet
* Images
* Documents
* Any file type

Features:
* Flexible storage.
* Data staging area.
* No schema enforcement.
* Cannot be queried directly with SQL.

Use for:
* Raw data ingestion
* Historical storage
* Pre-transformation staging
##### Common Workflow
```
Raw Files → Spark / Dataflows → Transform Data → Delta Tables → SQL / Power BI / AI
```
#### Delta Lake Tables
Delta Lake is the technology that powers lakehouse tables.
##### 1. ACID Transactions
Ensures:
* Consistency
* Reliability
* Concurrent reads and writes

Meaning multiple users can work with data safely.
##### 2. Schema Enforcement
Delta validates incoming data.

Benefits:
* Prevents invalid data loads.
* Maintains data quality.
##### 3. Time Travel
Delta keeps a transaction log.

Allows you to:
* Query previous versions.
* Recover from mistakes.
* Roll back changes.
##### 4. Efficient Updates and Deletes
Unlike traditional files:
* `UPDATE` supported
* `DELETE` supported
* `MERGE` supported

Without rewriting entire datasets.
##### Delta Architecture
Each Delta table consists of:
```
Parquet Data Files + Transaction Log
```
The transaction log records all changes and enables advanced capabilities.

### **Understand data warehouses**

#### What is a Data Warehouse?
A **data warehouse** is a centralized repository designed for:
* Analytics
* Reporting
* Business Intelligence (BI)

Unlike operational databases that support daily transactions, a data warehouse is optimized for analyzing large volumes of historical data.
##### Core Data Warehouse Process
> Data Sources → Data Ingestion → Data Storage → Data Processing → Analysis & Reporting

**Steps:**
1. **Data Ingestion** – Collect data from source systems.
2. **Data Storage** – Store data in an analytics-friendly structure.
3. **Data Processing** – Transform and prepare data.
4. **Analysis & Delivery** – Generate insights and reports.
#### Fact Tables and Dimension Tables
Dimensional modeling organizes warehouse data into:
##### Fact Tables
Contain measurable business events and metrics.

Examples:
* Sales transactions
* Revenue
* Orders
* Inventory movements

Characteristics:
* Large number of rows
* Numeric values for analysis
* Foreign keys to dimensions

Example:
| DateKey  | ProductKey | CustomerKey | SalesAmount |
| -------- | ---------- | ----------- | ----------- |
| 20240101 | 101        | 5001        | 250         |

##### Dimension Tables
Contain descriptive business information.

Examples:
* Customers
* Products
* Stores
* Dates

Characteristics:
* Fewer rows than fact tables
* Provide context for facts

Example:
| CustomerKey | CustomerName | Region |
| ----------- | ------------ | ------ |
| 5001        | John Smith   | West   |

##### Keys in Dimension Tables
**Surrogate Key**

A warehouse-generated key.

Characteristics:
* Usually an integer
* Unique within the warehouse
* Used for relationships

Example:
```
CustomerKey = 1001
```
**Alternate Key**

A business key from the source system.

Examples:
* `CustomerID`
* `ProductCode`
* `EmployeeNumber`

Example:
```
CustomerID = CUST-4582
```
##### Why Use Both?
| Key Type      | Purpose                                 |
| ------------- | --------------------------------------- |
| Surrogate Key | Warehouse relationships and consistency |
| Alternate Key | Traceability to source systems          |

#### Special Dimension Types
##### Time Dimension
Stores calendar information.

Typical columns:
```
Date
Day
Month
Quarter
Year
```
Enables analysis such as:
* Monthly sales
* Quarterly revenue
* Year-over-year trends
##### Slowly Changing Dimensions (SCD)
Track changes in attributes over time.

Examples:
* Customer address changes
* Product price changes
* Store location changes

Benefit:
* Preserves history
* Supports historical reporting
#### Schema Design
##### Normalization vs Denormalization
**Transaction Systems**

Typically **normalized**:
* Minimize duplicate data
* More tables
* More joins

**Data Warehouses**

Typically **denormalized**:
* Fewer joins
* Faster analytics
* Easier reporting
#### Star Schema
The most common data warehouse design.

Structure:
```
          DimCustomer
                |
DimProduct -- FactSales -- DimDate
                |
            DimStore
```
Characteristics:
* Fact table in the center
* Dimensions directly connected
* Simple and high-performing
* Preferred for BI reporting
##### Benefits
* Simpler queries
* Better performance
* Easy for users to understand
* Ideal for Power BI
#### Snowflake Schema
A more normalized dimensional design.
##### Example:
```
FactSales
    |
DimProduct
   / \
Category Supplier

DimCustomer
     |
DimGeography

DimStore
     |
DimGeography
```
##### Characteristics:
* Dimensions split into additional tables
* Reduces duplicate dimension data
* More joins required
##### Example
Instead of storing category information inside:
```
DimProduct
```
It is moved into:
```
DimCategory
```
Similarly:
```
DimCustomer
     → DimGeography

DimStore
     → DimGeography
```
#### Star Schema vs Snowflake Schema
| Feature         | Star Schema  | Snowflake Schema                 |
| --------------- | ------------ | -------------------------------- |
| Simplicity      | High         | Lower                            |
| Performance     | Faster       | Slightly slower                  |
| Number of joins | Fewer        | More                             |
| Normalization   | Denormalized | Partially normalized             |
| BI Reporting    | Preferred    | Used when dimensions are complex |

#### Exam Quick Facts
* Data warehouses are optimized for analytics and reporting
* Dimensional modeling uses Fact Tables and Dimension Tables
* Fact tables contain measurable business events
* Dimension tables contain descriptive attributes
* Surrogate keys are warehouse-generated identifiers
* Alternate keys are source-system business identifiers
* Time dimensions support date-based analysis
* Slowly Changing Dimensions (SCDs) preserve historical changes
* Star schema is the most common warehouse design
* Snowflake schema normalizes dimensions into additional tables
#### Memory Trick
* Facts = Numbers Dimensions = Descriptions
* Star = Simple & Fast Snowflake = More Normalized

#### Query data warehouse tables
Since the data warehouse is a relational database, you can use SQL to query its tables.
##### Query fact and dimension tables
Most queries in a relational data warehouse involve aggregating and grouping data (using aggregate functions and `GROUP BY` clauses) across related tables (using `JOIN` clauses).
1. Navigate back to the browser tab with your warehouse and create a new SQL Query, and run the following code:
    ```sql
    SELECT  d.[Year] AS CalendarYear,
            d.[Month] AS MonthOfYear,
            d.MonthName AS MonthName,
            SUM(so.SalesTotal) AS SalesRevenue
    FROM FactSalesOrder AS so
    JOIN DimDate AS d ON so.SalesOrderDateKey = d.DateKey
    GROUP BY d.[Year], d.[Month], d.MonthName
    ORDER BY CalendarYear, MonthOfYear;
    ```
Note that the attributes in the date dimension enable you to aggregate the measures in the fact table at multiple hierarchical levels - in this case, year and month. This is a common pattern in data warehouses.
2. Modify the query as follows to add a second dimension to the aggregation.
    ```sql
    SELECT  d.[Year] AS CalendarYear,
            d.[Month] AS MonthOfYear,
            d.MonthName AS MonthName,
            c.CountryRegion AS SalesRegion,
            SUM(so.SalesTotal) AS SalesRevenue
    FROM FactSalesOrder AS so
    JOIN DimDate AS d ON so.SalesOrderDateKey = d.DateKey
    JOIN DimCustomer AS c ON so.CustomerKey = c.CustomerKey
    GROUP BY d.[Year], d.[Month], d.MonthName, c.CountryRegion
    ORDER BY CalendarYear, MonthOfYear, SalesRegion;
    ```
3. Run the modified query and review the results, which now include sales revenue aggregated by year, month, and sales region.
#### Create a view
A data warehouse in Microsoft Fabric has many of the same capabilities you may be used to in relational databases. For example, you can create database objects like *views* and *stored procedures* to encapsulate SQL logic.
1. Modify the query you created previously as follows to create a view (note that you need to remove the `ORDER BY` clause to create a view).
    ```sql
    CREATE VIEW vSalesByRegion
    AS
    SELECT  d.[Year] AS CalendarYear,
            d.[Month] AS MonthOfYear,
            d.MonthName AS MonthName,
            c.CountryRegion AS SalesRegion,
            SUM(so.SalesTotal) AS SalesRevenue
    FROM FactSalesOrder AS so
    JOIN DimDate AS d ON so.SalesOrderDateKey = d.DateKey
    JOIN DimCustomer AS c ON so.CustomerKey = c.CustomerKey
    GROUP BY d.[Year], d.[Month], d.MonthName, c.CountryRegion;
    ```
Run the query to create the view. Then refresh the data warehouse schema and verify that the new view is listed in the **Explorer** pane.
3. Create a new SQL query and run the following SELECT statement:
    ```sql
    SELECT CalendarYear, MonthName, SalesRegion, SalesRevenue
    ```

##### Why KQL?
KQL is optimized for:
* Large datasets

* Time-series analytics
* Fast aggregations

KQL databases automatically:
* Index incoming data
* Partition data by ingestion time
* Optimize query performance
#### Basic KQL Syntax
KQL uses a pipe (`|`) syntax.

General pattern:
```KQL
TableName
| Operator
| Operator
| Operator
```
Think of each pipe as passing results to the next step.
##### Return Sample Records
```KQL
stock
| take 10
```
Returns any 10 rows from the table.
##### Filter Data
```KQL
stock
| where symbol == "MSFT"
```
Filters rows.
##### Aggregate Data
```KQL
stock
| where ["time"] > ago(5m)
| summarize avgPrice = avg(todouble(bidPrice)) by symbol
| project symbol, avgPrice
```
This query:
1. Looks at the last 5 minutes.
2. Calculates average price.
3. Groups by stock symbol.
4. Returns selected columns.

| --------- | ------------------------- |

### **Automate actions**

#### What is Activator?
**Activator** is a Microsoft Fabric capability that monitors real-time data and automatically triggers actions when specific conditions occur.

Instead of only visualizing data, Activator helps you respond automatically to events.
##### Examples
* Send an email when a temperature exceeds a threshold.
* Trigger a Power Automate workflow when sales drop.
* Run a notebook when data arrives.
* Alert operations teams when deliveries are delayed.
#### Activator Core Concepts
Activator is built around four key concepts:
##### 1. Events
An **event** is something that happened at a specific point in time.

Examples:
* Product sold
* Package delivered
* Sensor reading received
* Website error occurred

Remember

✅ Event = One occurrence at one moment in time
##### 2. Objects
An **object** is the business entity represented by events.

Examples:
* Sales order
* Customer account
* Delivery vehicle
* Temperature sensor

Example
```
Object = Delivery Truck
Events = GPS updates from the truck
```
##### 3. Properties
**Properties** are the fields or attributes associated with an object.

Examples:
```
Temperature
Order Total
Account Balance
Delivery Status
```
Example:
```
Object = Sensor
Property = Temperature
Value = 95°F
```
##### 4. Rules
A **rule** defines the condition that triggers an action.

Example:
```
IF Temperature > 100°
THEN Send Email Alert
```
Rules are the heart of Activator.

Formula
> Event → Object → Property → Rule → Action

The eventstream captures real-time stock market data and loads it into a table in your KQL database. You can query this table to see the captured data.

7. Wait a few seconds and run it again, noting that the average prices change as new data is added to the table from the real-time stream.
