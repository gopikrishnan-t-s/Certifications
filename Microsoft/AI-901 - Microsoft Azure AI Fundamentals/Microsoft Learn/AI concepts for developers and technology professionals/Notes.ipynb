* Understand language
* Generate content
* Recognize speech
* Analyze images
* Extract information
* Automate tasks

* Images
* Videos

* Conversations

# Introduction to generative AI and agents

## 1. Generative AI Overview
### Definition
Generative AI is a branch of AI that creates new content such as:
* Text
* Images
* Code
* Audio
* Videos
### Key Point
Generative AI appears intelligent because it learns patterns and relationships from large datasets and uses them to generate new content.

## 2. Large Language Models (LLMs)
### Definition
LLMs are the core technology behind generative AI. They learn linguistic and semantic relationships between words and phrases and generate meaningful responses.
### How LLMs Work
LLMs predict the next most likely token in a sequence, similar to predictive text on smartphones but much more advanced.
### Example
Prompt:
> "When my dog was a..."

Predicted completion:
> "puppy" is more likely than "skateboard" because of learned language relationships.

### Exam Definition
LLM = A model trained to predict and generate the next token in a sequence based on context.

## 3. Tokenization
### Definition
Tokenization is the process of breaking text into smaller units called tokens.
### Tokens Can Be
* Whole words
* Parts of words
* Punctuation marks
* Character sequences
### Example
Sentence:
> I heard a dog bark loudly at a cat

Tokens:
* I
* heard
* a
* dog
* bark
* loudly
* at
* cat
### Exam Tip
Token = Smallest language unit used by LLMs.

## 4. Transformer Architecture
### Purpose
A transformer learns relationships between tokens and helps generate meaningful responses.
### Two Main Components
#### Encoder
* Creates embeddings.
* Uses attention mechanisms.
* Learns contextual meaning.
#### Decoder
* Predicts the next token.
* Generates text responses.
### Exam Question
What are the two parts of a transformer? ✅ Encoder and Decoder

## 5. Embeddings
### Definition
Embeddings are vector representations of tokens that contain semantic and contextual meaning.
### Key Idea
Words with similar meanings have similar embeddings.

Examples:
* Dog
* Puppy
* Cat

These embeddings are closer together than:
* Dog
* Car
* Skateboard
### Importance
Embeddings help models understand meaning rather than simply memorizing words.

## 6. Attention Mechanism
### Definition
Attention determines which words in a sequence are most important when understanding a token.
### Example
Sentence:
> I heard a dog bark

Important words for "bark":
* heard
* dog

Less important:
* I
* a
### Multi-Head Attention
* Evaluates multiple relationships simultaneously.
* Improves efficiency and understanding.
### Exam Tip
Attention = Focus on relevant words in context.

## 7. Positional Encoding
### Problem
Transformers must know the order of words.
### Solution
Positional encoding adds information about each token's position in a sequence.
### Example
* Token: dog
* Position: 4

The model uses position information along with token information.

## 8. Predicting Text
### During Training
The model:
* Sees text sequences.
* Predicts missing tokens.
* Compares prediction with actual token.
* Adjusts weights to reduce errors.
### During Use
1. User enters prompt.
2. Model predicts next token.
3. Token is added.
4. Process repeats until completion.

## 9. Prompts
### Definition
A prompt is the input given to an LLM.
### Two Types of Prompts
#### System Prompt
Defines:
* Role
* Tone
* Behavior
* Constraints

Example:
> "You are a helpful assistant."

#### User Prompt
Specific instruction or question.

Example:
> "Summarize this document in 6 bullet points."

### Exam Tip
| Prompt Type | Purpose |
| --- | --- |
| System Prompt | Controls behavior |
| User Prompt | Requests task |

## 10. Conversation History
### Definition
Previous interactions included in future prompts to maintain context and consistency.
### Benefit
Allows meaningful follow-up questions without repeating information.

## 11. Retrieval-Augmented Generation (RAG)
### Definition
RAG retrieves external information and includes it in the prompt before generating a response.
### Process
1. User asks a question.
2. Relevant documents are retrieved.
3. Retrieved content is added to prompt.
4. LLM generates a context-aware answer.
### Benefits
* More accurate
* More relevant
* Grounded in actual documents
### Exam Definition
RAG = Retrieval + Prompt Augmentation + Generation

## 12. Prompt Engineering Best Practices
### 1. Be Clear and Specific
Avoid vague instructions.
### 2. Add Context
Include audience, topic, and purpose.
### 3. Use Examples
Show desired style/output.
### 4. Ask for Structure
Request:
* Bullet points
* Tables
* Numbered lists
### Easy Mnemonic
C-C-E-S
* Clear
* Context
* Examples
* Structure

## 13. AI Agents
### Definition
AI agents are software applications that:
* Reason using generative AI

* Respond to changing conditions
### Three Components
#### 1. LLM

### Definition
Multiple specialized AI agents working together.
### Example Workflow
* Agent 1 → Collect data
* Agent 2 → Analyze data
* Agent 3 → Execute action

Together they form an AI-powered workforce.
### Communication
Agents communicate through prompts and task delegation.

# Introduction to natural language processing concepts

## 1. Natural Language Processing (NLP)
### Definition
NLP is a branch of AI that enables computers to understand, analyze, and extract meaning from human language.
### Main Text Analysis Tasks
* Language Detection
* Key Term Extraction
* Entity Detection
* PII Detection
* Text Classification
* Sentiment Analysis
* Text Summarization

## 2. Language Detection
### Definition
Identifies the language(s) used in text.
### Example
Determining whether a document is:
* English
* French
* Hindi
* Mixed-language text
### Exam Point
Often the first step in a text processing workflow.

## 3. Key Term Extraction
### Definition
Finding important words or phrases that represent the main topics of a document.
### Example
Document about AI:
* AI
* Machine Learning
* Automation
* Productivity

## 4. Entity Detection (Named Entity Recognition - NER)
### Definition
Identifies named entities in text such as:
* People
* Places
* Dates
* Organizations
### Example
Sentence:
> Microsoft was founded by Bill Gates.

Entities:
* Microsoft → Organization
* Bill Gates → Person

## 5. PII Detection
### Definition
PII (Personally Identifiable Information) Detection identifies and redacts sensitive personal information.
### Examples
* Names
* Addresses
* Phone numbers
* Bank details
* Financial accounts
### Purpose
Privacy protection and regulatory compliance.

## 6. Text Classification
### Definition
Assigning text into predefined categories.
### Example
Email:
* Spam
* Not Spam

News:
* Sports
* Politics
* Technology

## 7. Sentiment Analysis
### Definition
A special type of text classification that determines emotional tone.
### Categories
* Positive
* Neutral
* Negative
### Applications
* Customer reviews
* Social media monitoring
* Product feedback analysis

## 8. Text Summarization
### Definition
Reducing text length while preserving key information.
### Example
Converting a 10-page report into a short summary.

## 9. Tokenization
### Definition
The process of breaking text into tokens (words, phrases, punctuation, etc.).
### Example
Sentence:
> We choose to go to the moon

Tokens:
* We
* choose
* to
* go
* to
* the
* moon
### Exam Definition
✅ Tokenization = Breaking text into smaller units called tokens

## 10. Text Preprocessing Techniques
### A. Text Normalization
#### Purpose
* Convert text to lowercase
* Remove punctuation
* Standardize text format

Example:
> AI, Ai, ai → ai

### B. Stop Word Removal
#### Definition
Removing words that carry little meaning.

Examples:
* the
* a
* it
* an
#### Benefit
Highlights more important words.
### C. N-Gram Extraction
#### Definition
Finding sequences of words.

| Type | Example |
| --- | --- |
| Unigram | AI |
| Bigram | Artificial Intelligence |
| Trigram | Natural Language Processing |

### D. Stemming
#### Definition
Removing word endings to obtain root forms.

Examples:
* Powered → Power
* Powering → Power
* Powerful → Power
#### Limitation
May not produce valid dictionary words.
### E. Lemmatization
#### Definition
Converts words into valid dictionary base forms (lemmas).

Examples:
* Running → Run
* Global → Globe
#### Difference
| Stemming | Lemmatization |
| --- | --- |
| Cuts endings | Uses language rules |
| May create invalid words | Creates valid words |
| Simpler | More accurate |

### F. POS Tagging
#### Definition
Assigning grammatical labels to words.

Examples:
* Noun
* Verb
* Adjective
* Adverb
#### Full Form
POS = Parts of Speech

## 11. Statistical Text Analysis
### Frequency Analysis
#### Definition
Counts how frequently terms appear in a document.
#### Assumption
Frequently occurring words often indicate the document's topic.
#### Example
Most frequent terms:
* AI
* Business
* Customer

Suggest document discusses AI in business.

## 12. TF-IDF
### Full Form
Term Frequency - Inverse Document Frequency
### Purpose
Finds words important to one document but uncommon across other documents.
### Formula Components
#### TF (Term Frequency)
Number of times a term appears.
#### IDF (Inverse Document Frequency)
Measures how rare a term is across all documents.
#### TF-IDF
$TF × IDF$
### Key Idea
| High TF-IDF | Low TF-IDF |
| --- | --- |
| Important in one document | Common in many documents |

### Exam Tip
Words appearing everywhere (AI, Microsoft, Agent) receive low discriminative value.

## 13. Bag-of-Words (BoW)
### Definition
Represents text as word occurrence or frequency vectors while ignoring:
* Grammar
* Word order
### Uses
* Spam detection
* Sentiment analysis
* Text classification
### Example
Words like:
* Miracle cure
* Lose weight fast

May indicate spam emails.

## 14. TextRank
### Definition
An unsupervised graph-based ranking algorithm used for text summarization and keyword extraction.
### Based On
Google's PageRank concept.
### Steps
#### 1. Build Graph
Each sentence becomes a node.
#### 2. Calculate Scores
Using sentence similarity.
#### 3. Select Highest Ranked Sentences
Used as summary.
### Applications
* Text summarization
* Keyword extraction

## 15. Semantic Language Models
### Definition
Modern NLP models that represent words as vectors (embeddings) containing semantic meaning.
### Examples
* Word2Vec
* GloVe
* GPT family models
### Benefit
Capture meaning rather than simple word counts.

## 16. Embeddings
### Definition
Numeric vector representations of words that encode semantic meaning.
### Example
Words:
* Dog
* Puppy
* Cat

Have similar vector directions.
### Exam Point
Semantically similar words → Similar vectors

## 17. Cosine Similarity
### Purpose
Measures similarity between vectors.
### Interpretation
| Value | Meaning |
| --- | --- | 
| Close to 1 | Very similar |
| Close to 0 | Less similar |

### Example
* Dog ↔ Cat = High similarity
* Dog ↔ Tree = Low similarity

## 18. Vector Arithmetic
### Concept
Word relationships can be modeled mathematically.

Examples:
* dog + young = puppy
* cat + young = kitten
* puppy - young = dog

### Exam Question
What NLP capability allows semantic reasoning through vector addition and subtraction? ✅ Vector Arithmetic

## 19. Analogical Reasoning
### Definition
Using vector operations to solve analogies.

Example:
> puppy : dog :: kitten : cat

Formula:
$kitten - puppy + dog = cat$

## 20. Applications of Semantic Models
### Text Summarization

### Keyword Extraction
Find important document terms.
### Named Entity Recognition
Identify persons, locations, organizations.
### Text Classification
Used in:
* Sentiment analysis
* Topic categorization
* Document classification

## 8. Speech Recognition Pipeline

## 9. Audio Capture
### Purpose
Convert analog speech into a digital signal.
### Typical Sampling Rate
* 16,000 samples per second
* 16 kHz
### Additional Processing
Noise filtering removes:
* Hums
* Clicks
* Background noise

## 10. Pre-processing
### Purpose
Extract useful speech features from raw audio.
### Main Technique
✅ MFCC (Mel-Frequency Cepstral Coefficients)

## 11. MFCC
### Full Form
Mel-Frequency Cepstral Coefficients
### Purpose
Represents audio using compact feature vectors that closely match human hearing.
### Steps
1. Divide audio into frames
2. Apply Fourier Transform
3. Convert frequencies to Mel scale
4. Extract coefficients

Usually: ✅ 13 coefficients per frame
### Exam Question
What is the most common feature extraction technique in speech recognition? ✅ MFCC

## 12. Acoustic Modeling
### Purpose
Convert MFCC features into phoneme predictions.
### Phoneme Definition
Smallest meaningful unit of sound.

Example:
```
cat = /k/ + /æ/ + /t/
```
### Technology Used
✅ Transformer Models
### Key Transformer Features
* Attention mechanism
* Context awareness

## 2. Major Computer Vision Tasks

#### Definition
Identifying the main object or subject in an image and assigning a label.
#### Example
Image → Banana

Output:

Banana
#### Characteristics
* One label per image
* Identifies what the image contains
### B. Object Detection
#### Definition
Detects multiple objects and their locations within an image.
#### Output
* Object name
* Bounding box coordinates

Example:
* Apple → Box 1
* Banana → Box 2
* Orange → Box 3

| Image Classification | Object Detection |
| --- | --- |

| No location | Finds locations |
| What is it? | What & Where? |

#### Definition
Classifies every pixel in an image according to the object to which it belongs.
#### Advantage
More precise than object detection because it identifies exact object boundaries.
#### Exam Tip
✅ Object Detection → Bounding Boxes

✅ Semantic Segmentation → Pixel-Level Detection

# Introduction to retrieval-augmented generation concepts

## 1. What is Retrieval-Augmented Generation (RAG)?
### Definition
**Retrieval-Augmented Generation (RAG)** combines search/retrieval with generative AI to produce accurate, grounded, and context-aware responses.
### Why RAG is Needed
LLMs have limitations:
* Knowledge boundaries (can't access private company data)
* Knowledge freshness (training data becomes outdated)
* Verifiability issues (may generate unsupported answers)
### Solution
RAG first retrieves relevant information and then supplies it to the LLM before generating a response.

## 2. Core Components of RAG
### Retrieval
Searches data sources for relevant information.
### Augmented Generation
Adds retrieved information to the prompt and generates a response using that context.
### Formula
```
Retrieval + Context + Generation = RAG
```

## 3. Grounded Responses
### Definition
A grounded response is an answer supported by retrieved source content rather than relying solely on model training data.
### Benefits
* More accurate
* More trustworthy
* Verifiable with citations
* Less hallucination risk
### Exam Definition
✅ Grounded Response = Answer based on retrieved evidence.

## 4. RAG Architecture Overview
```
User Question
      ↓
Retrieve Information
      ↓
Add Context
      ↓
LLM Generation
      ↓
Grounded Response
```

## 5. Knowledge Sources for RAG
RAG systems may retrieve information from:
* Product manuals
* Policy documents
* Web pages
* Databases
* Support tickets
* Internal documents
### Important Note
Users should only retrieve information they are authorized to access.

## 6. RAG Indexing Pipeline
Before retrieval can occur, data must be prepared and indexed.
### Main Steps
```
Source Data
      ↓
Text Extraction
      ↓
Chunking
      ↓
Embeddings
      ↓
Index Creation
```
### Exam Tip
✅ Memorize all four indexing steps.

## 7. Source Data Processing
### Key Activities
#### Load
Read content from sources.
#### Extract

#### Clean
Remove:
* Repeated headers
* Navigation text
* Irrelevant content
#### Add Metadata
Examples:
* Title
* Source URL
* Category
* Access permissions
* Last updated date

## 8. Chunking
### Definition
Splitting large documents into smaller sections called chunks.
### Why Chunking?
Documents are often too large to send directly to an LLM.
#### Example
Travel Policy:
* Flight Policy
* Hotel Policy
* Meal Policy
* Transportation Policy

Each section becomes a chunk.
### Chunking Trade-offs
#### Large Chunks
Pros:
* More context

Cons:
* More irrelevant information
* Larger prompts
#### Small Chunks
Pros:
* More focused retrieval

Cons:
* Context may be lost
### Common Practice
✅ Slight overlap between chunks.

## 9. Embeddings
### Definition
Embeddings are numerical vectors representing semantic meaning.
### Purpose
Convert each chunk into vector form.
### Key Concept
Chunks with similar meanings have embeddings located close together in vector space.

## 10. Indexing
### Definition
Store chunks, embeddings, and metadata in a searchable system.
### Search Methods
#### Keyword Search
Matches exact words.

Best for:
* Product codes
* IDs
* Exact terms
#### Vector Search
Uses semantic similarity.

Best for:
* Meaning-based retrieval
#### Hybrid Search
Combines keyword and vector search.
### Exam Question

✅ Hybrid Search

## 11. Retrieval Phase
Once data is indexed, users can ask questions.
### Process
```
User Prompt
      ↓
Search Index
      ↓
Retrieve Chunks
      ↓
Add Context
      ↓
Generate Response
```

## 12. Query Embeddings
### Process
The user question is converted into an embedding using the same embedding model used during indexing.
### Goal
Find stored chunks with the closest semantic meaning.

## 13. Retrieval Optimization Techniques
### Metadata Filtering
Examples:
* Product
* Region
* Date
* User permissions
### Hybrid Search
Combines vector and keyword search.
### Ranking
Places best results first.
### Result Limiting
Reduces irrelevant context.
### Key Principle
✅ Retrieve the smallest amount of information that sufficiently answers the question.

## 14. Prompt Augmentation
### Definition
Adding retrieved content into the prompt before sending it to the LLM.
### Prompt Components
#### System Instructions
Define model behavior.
#### User Question
Original query.
#### Retrieved Chunks
Grounding information.
#### Answering Instructions
Examples:
* Use only provided context.
* Cite sources.
* State when information is insufficient.

## 15. Prompt Security
### Important Rule
Retrieved documents must be treated as data, not instructions.
### Reason
Source content may contain instructions that attempt to manipulate model behavior.

## 16. Response Generation
### Goal
Generate answers using retrieved evidence.
### Best Practices
* Include citations
* Link sources
* Acknowledge insufficient information
* Avoid unsupported claims

## 17. RAG Evaluation
### Two Areas
```
Retrieval Evaluation + Generation Evaluation
```
### Important Exam Point
Good answers alone do not guarantee a good RAG system.

## 18. Retrieval Evaluation
### Purpose
Determine whether the search process finds the correct content.
### Criteria
#### Relevance
Does retrieved content answer the question?
#### Coverage
Is all required information retrieved?
#### Ranking
Are the best results first?
#### Security & Freshness
Is content authorized and up to date?

## 19. Generation Evaluation
### Purpose
Evaluate how well the LLM uses retrieved information.
### Criteria
#### Groundedness
Claims supported by sources.
#### Relevance
Actually answers the question.
#### Completeness
Contains required information.
#### Citation Quality
Citations support claims.
#### Appropriate Uncertainty
Avoids guessing when information is missing.

## 20. Test Dataset for RAG
Good evaluation datasets should contain:
* Realistic questions
* Expected source passages
* Expected answers
* Ambiguous questions
* No-answer questions
* Outdated content scenarios
* Unauthorized-access attempts

## 21. Quality vs Latency vs Cost
### Trade-off
More Retrieved Chunks:

✅ Better coverage

❌ Higher cost

❌ Slower responses

❌ Larger prompts

❌ More distractions for the model
### Goal
Balance:
* Quality
* Latency
* Cost
