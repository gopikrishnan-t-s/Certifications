## **3. Script Activity**
### Purpose
Execute SQL directly inside a pipeline.
### Supported Operations
#### Query
Returns data
```sql
SELECT *
FROM Table
```
#### Non-Query
No output
```sql
INSERT
UPDATE
DELETE
CREATE TABLE
DROP TABLE
```
#### Logging
Can log:
##### Activity Output
* Easy setup
* 4 MB limit
##### External Storage
* Unlimited logging
* Additional configuration required
### DP-600 Tip
Enable Script Logging when troubleshooting SQL activities.

## **4. Stored Procedure Activity**
### Purpose
Execute reusable SQL logic.
### Benefits
✅ Reuse business logic \
✅ Parameter passing \
✅ Aggregations \
✅ Data manipulation
### Example
```sql
sp_SavePublicationYearArticleCount
```
Calculates:
> Publication Year → Article Count

and stores result in a table.
### Summary
Use Stored Procedure Activity when business logic already exists inside SQL.

## **5. If Condition Activity (Very Important)**
### Purpose
Supports branching logic.
### Structure
```
IF TRUE
    Branch A

IF FALSE
    Branch B
```
### Expression Language
Uses:
```
@
```
Examples:
```
@equals()
@greater()
@less()
```
### Example
```
ColumnCount == 10
```
TRUE:
> Process Data

FALSE:
> Log Error

### DP-600 Tip
Know:
* `@equals`
* `@greater`
* `@less`

These appear frequently in pipeline expressions.

## **6. Pipeline Dynamic Content**
### Purpose
Access activity results programmatically.
### Example
```
activity('lookup')
.output.firstRow
```
### Example
```
activity('check-column-count')
.output.resultSets[0].rows[0]
```
### Summary
Dynamic content allows activities to pass results to subsequent activities.

## **7. ForEach Activity (Exam Favorite)**
### Purpose
Loop through collections.
### Use Cases
✅ Process records one-by-one \
✅ Process arrays \
✅ Batch processing
### Modes
#### Sequential
One record at a time.
#### Parallel
Process batches simultaneously.
### Configuration
```
@activity('group-data')
.output.resultSets[0].rows
```
Each row becomes one iteration.
### Example
```
Publication Year < 2000
        ↓
Pre2000 Table
```
```
Publication Year >= 2000
        ↓
Post2000 Table
```
### Summary
ForEach enables row-wise workflow execution within pipelines.

## **8. Incremental Loading (Very Important)**
### Purpose
Load only new data.
##### Avoid:
❌ Full reloads
##### Benefits:
✅ Faster execution \
✅ Reduced compute cost \
✅ Better scalability
### Components
#### Source Table
Contains data with timestamps.

Example:
```
InvoiceDate
```
#### Watermark Table
Stores:
```
Last Loaded Timestamp
```
### Incremental Load Logic
> Data Timestamp > Watermark Value

If TRUE:
> Copy New Data

Else:
> No Action

## **9. Lookup Activities**
### Purpose
Retrieve values from tables.
### Lookup 1
Retrieve:
```
WatermarkValue
```
from:
```
watermark_table
```
### Lookup 2
Retrieve:
```sql
MAX(InvoiceDate)
```
from:
```
online_retail
```
### Important Option
> First Row Only

Allows direct access:
> output.firstRow

### Summary
Lookup Activities drive incremental-load decisions.

## **10. Incremental Copy Activity**
### Purpose
Copy only new rows.
### Query Pattern
```sql
SELECT *
FROM online_retail
WHERE InvoiceDate > Watermark
AND InvoiceDate <= LatestTimestamp
```
### Destination
Lakehouse Files

Example:
```
IncrementalCopy/
```
Stored as:
```
CSV
```
### Summary
Copy Activity supports dynamic filtering for incremental data movement.

## **13. Delta Table Partitions (Important)**
### Purpose
Improve query performance.
### Example
Partition By:
```
Region
Category
```
Result:
```
Region
   ├─ Category
        ├─ Parquet Files
```
### Supported Partition Types
✅ String \
✅ Integer \
✅ Boolean \
✅ Datetime
### Benefits
✅ Smaller scans \
✅ Faster queries \
✅ Lower compute usage
### DP-600 Tip
Partition by columns frequently used in filters.

## **14. Dataflow Gen2 vs Data Pipelines**
| Feature           | Dataflow Gen2 | Data Pipeline |
| ----------------- | ------------- | ------------- |
| Power Query       | ✅             | ❌             |
| Transformation    | ✅             | ✅             |
| Orchestration     | ❌             | ✅             |
| Scheduling        | ❌             | ✅             |
| Loops             | ❌             | ✅             |
| If Conditions     | ❌             | ✅             |
| Notebook Support  | ❌             | ✅             |
| SQL Scripts       | ❌             | ✅             |
| Stored Procedures | ❌             | ✅             |

### When to Use Dataflow Gen2
✅ Data Cleaning \
✅ Data Preparation \
✅ Joining Data \
✅ No-Code Transformations
### When to Use Pipelines
✅ Orchestration \
✅ Scheduling \
✅ Automation \
✅ Conditional Logic \
✅ Incremental Loads \
✅ Running Code
