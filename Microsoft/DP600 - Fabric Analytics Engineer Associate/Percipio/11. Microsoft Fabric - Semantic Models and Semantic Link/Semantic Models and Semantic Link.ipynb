## **2. Creating a Semantic Model with a Lakehouse Table**
### Explanation
The course demonstrates creating Semantic Models directly from Delta Tables inside a Lakehouse.
### Process
#### Step 1
Upload data into Lakehouse Files.
#### Step 2
Load file into:
```
Delta Table
```
#### Step 3
Create Semantic Model.
#### Step 4
Build Power BI Reports.
### Default Storage Mode
A Lakehouse-based Semantic Model defaults to:

✅ Direct Lake Mode

because Delta Tables are stored in OneLake.

#### Direct Lake Behavior Options
| Option | Behavior |
| --- | --- |
| Automatic | Direct Lake first, fallback to DirectQuery |
| Direct Lake Only | No fallback |
| DirectQuery Only | Always query source |

### DP-600 Tip
* Automatic = Default Fabric Behavior
* Direct Lake Only = Highest Performance

## **3. Viewing Columns Stored in Memory in Direct Lake Mode**
### Explanation
One of the most important demonstrations in the course.
### Direct Lake Behavior
Only columns used by Power BI visuals are loaded into memory.
#### Example:
Power BI accesses:
* Country
* Profit
* OrderID

Direct Lake loads only these columns into memory.

### DMV Used
```sql
$SYSTEM.DISCOVER_STORAGE_TABLE_COLUMNS
```
Used to inspect:
* Memory usage
* Dictionary size
* Column residency
* Access frequency
### Important Metrics
#### `DICTIONARY_ISRESIDENT`
```
`True`  = In memory

`False` = Not loaded
```
#### `DICTIONARY_SIZE`
Memory consumed.
#### `DICTIONARY_TEMPERATURE`
How frequently column is accessed.
#### `DICTIONARY_LAST_ACCESSED`
Last load timestamp.
### Summary
Direct Lake loads only required columns into memory, providing Import-like performance without importing the entire dataset.

## **4. Direct Lake vs DirectQuery Memory Behavior**
### Explanation
The course compares memory management.
### Direct Lake
When a column is accessed:

✅ Stored in memory
#### Example:
* Country
* Profit
* OrderID

all resident in memory.
### DirectQuery
When queried:

✅ Data retrieved directly from source

❌ Columns may not become memory resident

Some categorical columns may appear in memory.

Many numeric columns remain source-bound.
### DP-600 Exam Memory Trick
| Feature | Direct Lake | DirectQuery |
| --- | --- | --- |
| In-Memory Columns | ✅ | Limited |
| Real-Time | ✅ | ✅ |
| Query Speed | Faster | Slower |
| Delta Tables | ✅ | Via SQL Endpoint |

## **6. Security for Semantic Models**
### Explanation
Three key security layers are discussed.
### Row-Level Security (RLS)
Restricts:
* Rows

Users see only their data.
#### Implemented using:
* Roles
* DAX filters
#### Example:
```dax
USERPRINCIPALNAME()
```
#### Important Rule
RLS applies only to:

✅ Viewers

Not:

❌ Admins \
❌ Members \
❌ Contributors
### Object-Level Security (OLS)
Restricts:
* Tables
* Columns
* Measures

Creates hidden objects.
### Column-Level Security (CLS)
Specialized OLS implementation.

Used for:
* Salary
* PII
* Financial data
### OLS Tools
Cannot be configured directly in Power BI Desktop.

Uses:
* Tabular Editor
* SSMS
* Visual Studio
### Summary
| Security Type | Restricts |
| --- | --- |
| RLS | Rows |
| CLS | Columns |
| OLS | Tables + Columns + Measures |

## **10. Evaluating Measures Using SemPy**
### Explanation
#### List Measures
```python
fabric.list_measures()
```
Shows all semantic model measures.
#### Evaluate Measure
```python
fabric.evaluate_measure()
```
Runs semantic model calculations directly from notebooks.
#### Grouped Evaluation
Example:

returns measure by year.
#### Filtering
Example:
```python
filters={
'DimTime[Year]' : ['2023']
}
```
returns filtered measure values.
### Summary
SemPy allows semantic model measures to be evaluated programmatically.
