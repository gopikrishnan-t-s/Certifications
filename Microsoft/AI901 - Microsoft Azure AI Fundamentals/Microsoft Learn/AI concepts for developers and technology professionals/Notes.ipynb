# Introduction to AI Concepts

## 1. Artificial Intelligence (AI)
### Definition
AI enables computers to perform tasks that normally require human intelligence. This module focuses on understanding AI capabilities rather than coding or mathematical details.
### Key Exam Point
AI is used to:
* Understand language
* Generate content
* Recognize speech
* Analyze images
* Extract information
* Automate tasks

## 2. Generative AI
### Definition
Generative AI creates new content such as:
* Text
* Images
* Videos
* Code
* Conversations
### How It Works
1. Trained on large amounts of data.
3. User provides a prompt.
3. Language model generates a meaningful response.
### LLM vs SLM
| LLM (Large Language Model) | SLM (Small Language Model) |
| --- | --- |
| Trained on huge datasets | Smaller datasets |
| More powerful | More focused |
| General-purpose | Specialized tasks |
| Higher cost | Lower cost |
| Better generalization | Easier local deployment |
### Common Uses
* Chatbots
* AI assistants
* Content creation
* Translation
* Summarization

## 3. AI Agents
### Definition
AI agents are software applications that:
* Reason using AI
* Use tools
* Automate tasks
* Respond to context
### Three Components of an Agent
1. Large Language Model (Brain)

    Provides language understanding and reasoning.

2. Instructions

    Defines the agent's role and behavior.

3. Tools

    Allow interaction with the world. Examples:
    * Search engines
    * Databases
    * Email systems
    * Calendars
    * Devices
### Exam Tip
Agent = LLM + Instructions + Tools ✅

## 4. Natural Language Processing (NLP)
### Definition
NLP is a field of AI focused on understanding and processing human language. It forms the foundation of modern generative AI systems.
### Text Analysis Techniques
#### Language Detection
Identifies the language used in text.
#### Text Classification
Assigns text into categories.
#### Sentiment Analysis
Determines whether text is:
* Positive
* Negative
* Neutral
#### Key-Term Extraction
Finds important words and phrases.
#### Entity Detection
Identifies:
* People
* Places
* Organizations
#### PII Detection
Finds personal information such as:
* Names
* Addresses
* Phone numbers
#### Summarization
Produces shorter versions while retaining key ideas.
#### Applications
* Call transcript analysis
* Meeting summaries
* Social media analysis
* Chatbots
* Privacy protection

## 5. Speech AI
### Definition
Allows users to interact using spoken language.
### Speech Recognition
Speech-to-Text (STT)
* Converts spoken words into text.
### Speech Synthesis
Text-to-Speech (TTS)
* Converts text into spoken audio.
### Applications
* Voice assistants
* Call transcription
* Audio descriptions
* Speech translation

## 6. Computer Vision
### Definition
AI technology used to analyze:
* Photos
* Videos
* Camera feeds
### Types of Computer Vision Models
#### Image Classification
Determines the main object or subject in an image.
#### Object Detection
Identifies objects and their locations.
#### Semantic Segmentation
Identifies exact pixels belonging to objects.
#### Multimodal Models
Combine images and text to generate descriptions.
### Applications
* Visual search
* Retail inventory tracking
* Security monitoring
* Facial recognition
* Robotics
* Self-driving vehicles

## 7. Information Extraction
### Definition
Extracting useful information from:
* Documents
* Forms
* Images
* Audio
* Video
### OCR (Optical Character Recognition)
#### Purpose
Detects and reads text from images.
#### Applications
* Expense claim processing
* Form digitization
* Document indexing
* Meeting action extraction
### Exam Tip
OCR = Image/Text Recognition Technology

## 8. Responsible AI
### Definition
Building AI systems with safeguards to reduce:
* Harmful content
* Illegal actions
* Offensive outputs

## 9. Six Principles of Responsible AI
### 1. Fairness
Avoid bias and discrimination.
### 2. Reliability and Safety
AI should operate safely and consistently.
### 3. Privacy and Security
Protect personal and organizational data.
### 4. Inclusiveness
Ensure AI can be used by everyone.
### 5. Transparency
Explain how AI works and its limitations.
### 6. Accountability
Organizations remain responsible for AI outcomes.
### Important Examples for Exams
| Scenario | Responsible AI Principle |
| --- | --- |
| Fair college admissions | Fairness |
| Robot avoids acting on low-confidence detections | Reliability & Safety |
| Delete temporary facial images | Privacy & Security |
| Speech + captions for hearing-impaired users | Inclusiveness |
| Explain AI use in loan approvals | Transparency |
| Organization responsible for AI decisions | Accountability |
