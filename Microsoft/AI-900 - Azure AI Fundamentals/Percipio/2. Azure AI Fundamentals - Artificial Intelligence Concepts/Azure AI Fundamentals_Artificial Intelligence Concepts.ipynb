## **Machine Learning (ML) Types**
### 1. What is Machine Learning (ML)?
Machine Learning is a core subset of Artificial Intelligence (AI) that enables systems to learn from data, identify patterns, and improve performance without being explicitly programmed.

Different ML approaches are designed for different types of problems and data.

### 2. Types of Machine Learning
#### A. Supervised Learning
**Definition**

Supervised learning trains a model using labeled data, where each input already has a known correct output.

##### How It Works
> Input + Known Answer → Training → Model → Prediction on New Data

##### Example
Predicting a car's price based on:
* Make
* Model
* Year
* Mileage

The model learns from historical car data where prices are already known.

##### Common Uses
* Classification
* Regression
* Forecasting

##### Examples
* Spam detection
* House price prediction
* Fraud detection

#### B. Unsupervised Learning
##### Definition
Unsupervised learning works with unlabeled data and discovers hidden patterns or structures.
##### How It Works
The system identifies similarities without being told the correct answers.
##### Example
A parking lot contains:
* Cars
* Trucks
* Motorcycles

The AI groups similar vehicles together without knowing their labels beforehand.

##### Common Uses
* Clustering
* Pattern discovery
* Customer segmentation
* vAnomaly detection

#### C. Semi-Supervised Learning
##### Definition
Uses a combination of:
* Small amount of labeled data
* Large amount of unlabeled data

##### Why It's Useful
* Labeling data is time-consuming and expensive.
* Semi-supervised learning improves accuracy while reducing labeling effort.

##### Benefits
✅ Lower cost \
✅ Better model performance \
✅ Efficient use of available data
##### Example
Image classification where only a portion of images are labeled.

#### D. Self-Supervised Learning
##### Definition
Uses unlabeled data to automatically create its own learning labels (pseudo-labels).
##### Key Idea
The model learns from the structure of the data itself, without human-provided labels.
##### How It Works
* Generates internal supervisory signals.
* Predicts missing or hidden information.
* Learns useful representations automatically.

##### Benefits
✅ Reduces reliance on labeled data \
✅ Scales well to large datasets \
✅ Widely used in modern AI systems
##### Examples
* Large Language Models (LLMs)
* GPT-style models
* Modern NLP systems

#### E. Reinforcement Learning
##### Definition
A model learns through trial and error by interacting with an environment.
##### Learning Process
* Perform action
* Receive reward or penalty
* Adjust behavior
* Improve future decisions
##### Key Components
| Component | Description |
| --- | --- |
| Agent | Learner/decision-maker |
| Environment | Surrounding world |
| Action | Decision made |
| Reward | Positive feedback |
| Penalty | Negative feedback |

##### Example
Teaching an AI to drive a car:

✅ Stay on track → Reward \
❌ Crash into wall → Penalty

The AI gradually learns the best driving strategy.

##### Applications
* Self-driving vehicles
* Robotics
* Game AI
* Autonomous systems

### 3. Quick Comparison of ML Types
| Learning Type | Data Used | Goal | Example |
| --- | --- | --- | --- |
| Supervised | Labeled | Predict known outcomes | House price prediction |
| Unsupervised | Unlabeled | Discover patterns | Customer segmentation |
| Semi-Supervised | Mixed | Improve learning with limited labels | Image classification |
| Self-Supervised | Unlabeled | Generate own labels | GPT/LLMs |
| Reinforcement | Rewards & Penalties | Learn optimal actions | Self-driving cars |

## **Labeled and Unlabeled Data**
### 1. Why Labeled and Unlabeled Data Matter
Data is the foundation of Machine Learning, and understanding the difference between **labeled** and **unlabeled** data is essential for selecting the right learning approach.

These data types are used in:
* Classification
* Clustering
* Pattern Recognition
* Computer Vision
* Natural Language Processing (NLP)

### 2. Labeled Data
#### Definition
**Labeled data** contains input data along with the correct output (label or tag).

Each record is assigned to a specific category or class.

##### Example
| Image | Label |
| --- | --- |
| 🐶 | Dog |
| 🐱 | Cat |
| 🚗 | Car |

The model learns from these known examples.

#### Characteristics of Labeled Data
##### Explicit Annotations
Each data instance has a known label.
##### Ground Truth
Labels represent the correct answer that the model should learn.
##### Multiple Labels Possible
A single record can belong to more than one category.

**Example**
| Object | Label 1 | Label 2 |
| --- | --- | --- |
| Person wearing red shirt | Person | Red |
| Blue Robot | Robot | Blue |

This is called **multi-label classification**.

#### Challenges of Labeled Data
##### Expensive
Requires human effort to create labels.
##### Time-Consuming
Large datasets can take significant time to annotate.
##### Specialized Knowledge
Certain domains require experts:
##### Examples:
* Medical imaging
* Legal documents
* Scientific research

#### Uses of Labeled Data
##### Supervised Learning
Models learn from input-output pairs.
##### Examples:
* Spam detection
* House price prediction
* Fraud detection

##### Semi-Supervised Learning
Uses:
* Small amount of labeled data
* Large amount of unlabeled data

##### Benefits:
✅ Lower labeling costs \
✅ Improved accuracy

### 3. Unlabeled Data
#### Definition
Unlabeled data contains data without any tags, labels, or predefined categories.
##### Example
A dataset of customer records:
| Customer |
| --- |
| Record A |
| Record B |
| Record C |

No information is provided about categories or outcomes.

#### Characteristics of Unlabeled Data
##### No Explicit Labels
The system receives only raw data.
##### Abundant
Much easier and cheaper to collect.
##### Pattern Discovery
The algorithm must identify patterns itself.

#### Uses of Unlabeled Data
**Unsupervised Learning**

Common tasks:
##### Clustering
Group similar data together.

##### Example:
> Customers → \
> Group 1 \
> Group 2 \
> Group 3

##### Dimensionality Reduction
Reduce complexity while maintaining important information.
##### Anomaly Detection
Find unusual or suspicious observations.
##### Examples:
* Fraud detection
* Network intrusion detection

### 4. Labeled vs Unlabeled Data
| Feature | Labeled Data | Unlabeled Data |
| --- | --- | --- |
| Contains Labels | Yes | No |

### 5. Importance of Data Labeling Quality
Machine Learning models are only as good as the data they learn from.

#### Good Labels Should Be
✅ Accurate \
✅ Consistent \
✅ Reliable \
✅ Representative of real-world scenarios

#### Labeling Quality Factors
##### Accuracy
Labels correctly reflect reality.

Example:
> Dog image → Dog

Not:
> Dog image → Cat

##### Consistency
Similar items receive the same labels.

Example: All invoices should be labeled as "Invoice."

Not:
> Invoice \
> Bill \
> Receipt

for the same type of document.

#### Real-World Relevance
Labels should reflect actual business or operational scenarios.

### 8. Intersection over Union (IoU)
#### What is IoU?
A metric used in **Computer Vision** to measure how well a predicted object location matches the actual object location.

**Formula**
$$\text{IoU} = \frac{\text{Overlap Area}}{\text{Combined Area}}$$

##### Interpretation
| IoU Score | Meaning |
| --- | --- |
| 1.0 | Perfect match |
| High | Good prediction |
| Low | Poor prediction |

##### Used For
* Object Detection
* Image Segmentation
* Bounding Box Evaluation

## **Data Features**
### 1. What are Data Features?
**Data features** are the measurable properties, attributes, or characteristics of data that machine learning models use to learn patterns and make predictions.

#### Characteristics of Good Features
✅ Descriptive – Capture important information about the data. \
✅ Distinctive – Differentiate between observations. \
✅ Independent – Minimize redundancy and unnecessary correlation with other features.

##### Examples
| Problem | Features |
| --- | --- |
| Car Price Prediction | Brand, Mileage, Horsepower, Age |
| House Price Prediction | Square Footage, Bedrooms, Location |
| Employee Attrition | Experience, Salary, Department |

### 2. Types of Data Features
#### Numeric Features
Contain numerical values.
##### Examples:
* Age
* Salary
* Temperature
* Horsepower

#### Categorical Features
Contain categories or labels.
##### Examples:
* Brand Name
* Country
* Product Type

#### Text/String Features
Contain textual information.
##### Examples:
* Customer reviews
* Emails
* Comments

#### Graph Features
Represent relationships and network structures.
##### Examples:
* Social networks
* Fraud detection networks
* Recommendation systems

### 3. Why Feature Selection Matters
Feature selection helps identify the most useful features for model training.
#### Benefits
✅ Improves model accuracy \
✅ Reduces training time \
✅ Reduces overfitting \
✅ Improves model interpretability \
✅ Removes irrelevant data \
✅ Improves generalization

### 4. Feature Selection Methods
#### A. Filter Methods
##### Definition
Evaluate features using statistical measures before model training.
##### Example
In a car-price dataset:
* Brand → Strong relationship to price ✅
* Paint Code → Weak relationship ❌

The algorithm keeps the most informative features.

##### Advantages
* Fast
* Computationally efficient
* Works well with large datasets

#### B. Wrapper Methods
##### Definition
Evaluate feature subsets by repeatedly training and testing models.
##### Process
1. Select feature subset
2. Train model
3. Evaluate performance
4. Repeat
##### Goal
Find the feature combination that produces the best model performance.
##### Advantage
Higher accuracy than filter methods.
##### Disadvantage
More computationally expensive.

#### C. Hybrid Methods
##### Definition
Combine:
* Statistical analysis (Filter)
* Model evaluation (Wrapper)
##### Benefit
Balancing speed and accuracy.

### 5. Advanced Feature Selection Methods
#### Embedded Methods
##### Definition
Feature selection happens naturally during model training.
##### Example
A model automatically discovers that:
* Brand
* Horsepower
* Vehicle Age

are significant predictors of price.

##### Advantage
Optimizes feature selection and model training simultaneously.
#### Optimization-Based Approaches
Use advanced algorithms such as:
* Genetic Algorithms
* Particle Swarm Optimization
* Evolutionary Strategies
##### Purpose
Efficiently explore feature combinations and identify optimal feature subsets.

### 6. Handling Multi-Collinearity
#### What is Multi-Collinearity?
Occurs when features are highly correlated with one another.
##### Example
| Feature |
| --- |
| Horsepower |
| Engine Size |

These may convey very similar information.
##### Problems
* Harder model interpretation
* Reduced reliability
* Unstable coefficient estimates
##### Solution
Feature selection removes redundant variables while keeping the most informative ones.

### 7. Statistical Significance & P-Values
##### Purpose
Determine whether a feature meaningfully contributes to predictions.
##### Example
Features:
* Brand
* Horsepower
* Color

A statistical test may reveal: \
✅ Brand → Significant \
✅ Horsepower → Significant \
❌ Color → Not significant

The insignificant feature may be removed.

### 8. Feature Selection Techniques
#### Forward Selection
##### Process
* Start with zero features.
* Add the most useful feature one at a time.
> No Features → Add Best Feature → Evaluate → Add Next Best Feature

##### Advantage
Simple and efficient.

#### Backward Elimination
##### Process
* Start with all features.
* Remove the least useful feature at each step.
> All Features → Remove Worst Feature → Evaluate → Repeat

##### Advantage
Works well when many features exist.

#### Recursive Feature Elimination (RFE)
##### Process
1. Train model
2. Rank feature importance
3. Remove least important feature
4. Retrain model
5. Repeat
##### Goal
Find the optimal feature subset.

### 9. Feature Importance Using Ensemble Models
Certain algorithms automatically calculate feature importance.
##### Random Forest
Measures how much each feature contributes to prediction accuracy.
##### Example
Most Important Features:
1. Brand
2. Mileage
3. Horsepower

#### XGBoost / LightGBM
Advanced boosting algorithms that: \
✅ Rank feature importance \
✅ Reduce overfitting \
✅ Improve prediction accuracy \
✅ Automatically identify useful predictors

### 10. Feature Selection and Bias-Variance Trade-Off
Feature selection helps balance:
#### Bias
Occurs when:
* Model is too simple
* Important features are missing
##### Result
Underfitting

#### Variance
Occurs when:
* Model is too complex
* Too many unnecessary features exist
##### Result
Overfitting

#### Goal
Select only the most informative features to achieve: \
✅ Better generalization \
✅ Better performance \
✅ Lower overfitting risk

## **Regression**
### 1. What is Regression?
**Regression** is a supervised machine learning technique used to predict continuous numeric values.

#### Answers Questions Like:
* How much?
* How many?
* What value?

##### Examples
* Predicting house prices
* Forecasting sales revenue
* Estimating stock prices
* Predicting vehicle costs

### 2. How Regression Works
Regression models learn the relationship between:
#### Inputs (Features)
Variables used to make predictions.

##### Examples:
* Mileage
* Horsepower
* House size
* Advertising spend

#### Output (Label)
A continuous value.

##### Examples:
* Car price
* House price
* Revenue
* Temperature

#### Process
> Features (Inputs) → Train Regression Model → Learn Relationships → Predict Continuous Value

### 3. Regression Training Process
Regression is a **supervised learning** technique, meaning it uses labeled data.

#### Step 1: Prepare Labeled Data
Example:
| Mileage | Age | Price |
| --- | --- | --- |
| 20,000 | 2 | $25,000 |
| 40,000 | 4 | $18,000 |

The correct answers (prices) are known.

#### Step 2: Split the Dataset
##### Training Dataset
Used to teach the model.
##### Validation Dataset
Used to test model performance during training.

> Full Dataset → Training Set → Validation Set

#### Step 3: Train the Model
The algorithm:
* Learns patterns
* Adjusts parameters
* Minimizes prediction errors

This process repeats over multiple iterations.

#### Step 4: Validate the Model
The trained model predicts values for validation data.

Predictions are compared against known answers.

$$\text{Predicted Price VS Actual Price}$$

#### Step 5: Evaluate Accuracy
Measure how close predictions are to actual values.

### 4. Regression Characteristics
#### Continuous Output
Unlike classification, regression predicts a value on a spectrum.
##### Examples
✅ House price = ₹58,75,000 \
✅ Sales forecast = ₹1,25,000 \
✅ Temperature = 33.5°C

#### Relationship Analysis
Regression helps understand how variables influence outcomes.

##### Example:
> Advertising Spend ↑ Sales Revenue

Higher advertising often increases revenue.

### 5. Common Regression Algorithms
#### A. Linear Regression
##### Definition
Models relationships using a straight-line equation.
##### Formula
$$\text{Y} = \text{MX + B}$$
##### Benefits
* Easy to interpret
* Fast
* Common AI-900 topic
##### Example
House size → House price

#### B. Polynomial Regression
##### Definition
Extends linear regression for curved relationships.
##### Use Case
When data does not follow a straight line.
##### Example:
> Speed vs Fuel Efficiency

#### C. Decision Tree Regression
##### Definition
Splits feature space into segments.
##### Prediction
Uses average target value within each segment.
##### Benefits
* Handles nonlinear relationships
* Easy to visualize

#### D. Neural Network Regression
##### Definition
Uses multiple layers of artificial neurons.
##### Benefits
* Captures complex patterns
* Handles large datasets
* Models nonlinear relationships

#### E. LASSO Regression
##### Full Form
*Least Absolute Shrinkage and Selection Operator*
##### Purpose
* Performs feature selection
* Reduces overfitting
##### Benefit
Removes less important features.

#### F. Ridge Regression
##### Purpose
Penalizes large regression coefficients.
##### Benefits
* Reduces overfitting
* Improves stability
* Handles multicollinearity

#### G. Elastic Net Regression
##### Definition
Combines:
* LASSO
* Ridge
##### Benefits
* Feature selection
* Regularization
* Improved performance

#### H. Stepwise Regression
##### Definition
Adds or removes features automatically.
##### Goal
Find optimal feature combinations.

#### I. Logistic Regression
##### Note for AI-900
Despite the name, Logistic Regression is mainly used for classification, not continuous regression.
##### Example
* Fraud or Not Fraud
* Yes or No
* Pass or Fail

### 6. Choosing a Regression Model
Selection depends on:
#### Data Characteristics

#### Prediction Goals

#### Computational Requirements

### 9. Regression Evaluation Metrics
#### A. Mean Absolute Error (MAE)
##### Definition
Average absolute difference between predictions and actual values.
##### Formula
$$\text{MAE} = \text{Average(|Actual - Predicted|)}$$
##### Benefits
* Easy to understand
* Treats all errors equally

#### B. Mean Squared Error (MSE)
##### Definition
Average squared difference between actual and predicted values.
##### Formula
$$\text{MSE} = \text{Average((Actual - Predicted)²)}$$
##### Benefits
* Penalizes large errors more heavily

#### C. Root Mean Squared Error (RMSE)
##### Definition
Square root of MSE.
##### Benefits
* Same unit as target variable
* Easier to interpret

#### D. R² (Coefficient of Determination)
##### Definition
Measures how much of the outcome variation is explained by the model.
##### Interpretation
| R² Value | Meaning |
| --- | --- |
| 1.0 | Perfect Fit |
| Close to 1 | Strong Model |
| Close to 0 | Weak Model |
##### Purpose
Evaluate overall model fit.

### 5. Classification Evaluation Metrics
To measure model performance, we compare predictions against actual values.
#### Confusion Matrix
A confusion matrix summarizes predictions.
|  | **Actual Positive** | **Actual Negative** |
| --- | --- | --- |
| **Predicted Positive** | True Positive (TP) | False Positive (FP) |
| **Predicted Negative** | False Negative (FN) | True Negative (TN) |
#### Accuracy
##### Definition
Percentage of correct predictions.
##### Formula
$$\text{Accuracy} = \frac{\text{(TP + TN)}}{\text{Total Predictions}}$$
##### Best When
Classes are balanced.

#### Precision
##### Definition
Measures how many predicted positives were actually positive.
##### Formula
$$\text{Precision} = \frac{\text{TP}}{\text{TP + FP}}$$
##### High Precision Means
Few false positives.

#### Recall (Sensitivity)
##### Definition
Measures how many actual positives were correctly identified.
##### Formula
$$\text{Recall} = \frac{\text{TP}}{\text{TP + FN}}$$
##### High Recall Means
Few false negatives.

#### Specificity
##### Definition
Measures how many actual negatives were correctly identified.
##### Formula
$$\text{Specificity} = \frac{\text{TN}}{\text{TN + FP}}$$
##### High Specificity Means
The model is good at identifying negatives.

#### F1 Score
##### Definition
Balances Precision and Recall.
##### Formula
$$\text{F1} = \frac{2\text{(Precision × Recall)}}{\text{(Precision + Recall)}}$$
##### Use Case
When both false positives and false negatives matter.

#### ROC Curve
##### Definition
Shows model performance across different classification thresholds.
##### Plots:
* True Positive Rate (Recall)
* False Positive Rate

#### AUC (Area Under Curve)
##### Definition
Measures overall classification performance.
##### Interpretation
| AUC Value | Meaning |
| --- | --- |
| 1.0 | Perfect Classifier |
| 0.9+ | Excellent |
| 0.8+ | Good |
| 0.5 | Random Guessing |

Higher AUC = Better classifier.

## **Clustering**
### 1. What is Clustering?
**Clustering** is an **unsupervised machine learning** technique used to group similar data points together without predefined labels.
#### Key Idea
* No known answers or labels exist.
* The algorithm discovers natural patterns and groupings within the data.
##### Example
A parking garage contains:

* Trucks
* Motorcycles
* Bicycles

### 2. Clustering and Unsupervised Learning
#### Characteristics
* Uses unlabeled data
* Discovers hidden patterns
* Finds similarities between data points
* Supports exploratory data analysis
#### Goal
Create **clusters** where:
* Data points within a cluster are similar.
* Data points in different clusters are dissimilar.

### 3. What is a Cluster?
A **cluster** is a collection of data points that share similar characteristics.
#### Good Clustering
* High similarity within clusters
* Low similarity between clusters
##### Example
> Cluster 1 → Cars \
> Cluster 2 → Trucks \
> Cluster 3 → Motorcycles

### 4. Types of Clustering Approaches
#### A. Centroid-Based Clustering
##### Definition
Data points are grouped around central points called centroids.
##### Goal
* Minimize distance within clusters.
* Maximize distance between clusters.
##### Example Algorithm
✅ K-Means

#### B. Density-Based Clustering
##### Definition
Clusters are identified as regions with high concentrations of data points.
##### Benefits
* Handles irregular cluster shapes
* Detects outliers
* Handles noise effectively
##### Example Algorithm
✅ DBSCAN

#### C. Distribution-Based Clustering
##### Definition
Assumes data follows probability distributions.
##### Goal
Assign data points to clusters based on probability.
##### Example Algorithm
✅ Gaussian Mixture Model (GMM)

#### D. Hierarchical Clustering
##### Definition
Builds a hierarchy of nested clusters.
##### Output
Creates a tree-like structure called a dendrogram.
##### Benefits
* Reveals cluster relationships
* Useful for exploratory analysis

### 5. Popular Clustering Algorithms
#### A. K-Means Clustering
**Most Important for AI-900**
##### How It Works
1. Choose number of clusters (`K`)
2. Assign data points to nearest centroid
3. Recalculate centroids
4. Repeat until clusters stabilize
##### Example
`K = 3`
##### Produces:
* Cars
* Trucks
* Motorcycles
##### Benefits
* Fast
* Easy to understand
* Widely used
##### Limitation
Requires predefined number of clusters.

#### B. DBSCAN
##### Full Form
*Density-Based Spatial Clustering of Applications with Noise*
##### How It Works
Groups dense regions together and identifies sparse regions as noise.
##### Benefits
* Handles outliers
* Finds arbitrary cluster shapes
* No need to specify cluster count

#### C. Gaussian Mixture Model (GMM)
##### How It Works
Assumes data is generated from a mixture of Gaussian distributions.
##### Benefits
* Probabilistic clustering
* Flexible cluster shapes

#### D. BIRCH
##### Full Form
*Balanced Iterative Reducing and Clustering using Hierarchies*
##### Purpose
Efficient clustering for very large datasets.
##### Benefits
* Memory efficient
* Scalable
* Hierarchical structure

#### E. Affinity Propagation
##### How It Works
Data points exchange messages and identify representative points (exemplars).
##### Benefits
* Automatically determines cluster centers
* No need to specify K

#### F. Mean Shift
##### How It Works
Moves data points toward the highest density regions.
##### Benefits
* Finds natural cluster centers
* No predefined cluster count required

#### G. OPTICS
##### Full Form
*Ordering Points To Identify the Clustering Structure*
##### Type
Density-based clustering.
##### Benefits
* Handles varying densities
* Reveals underlying cluster structure
* Improved flexibility over DBSCAN

#### H. Agglomerative Hierarchical Clustering
##### How It Works
1. Start with individual points
2. Merge closest clusters repeatedly
3. Build hierarchy
##### Output
Produces a dendrogram.
##### Benefits
* Visual cluster relationships
* Good for smaller datasets

### 2. What are Artificial Neural Networks (ANNs)?
An Artificial Neural Network (ANN) is a computational model made up of interconnected nodes (neurons) organized into layers.
#### Basic Structure
> Input Layer → Hidden Layer(s) → Output Layer

#### Components
##### Input Layer
Receives feature values.
##### Examples:
* Age
* Salary
* Image Pixels
* Sensor Data

##### Hidden Layers
Perform calculations and learn patterns.
* May contain one or many layers.
* More hidden layers = deeper network.

##### Output Layer
Produces final predictions.
##### Examples:
* Cat or Dog
* House Price
* Fraud or Not Fraud

### 3. Deep Neural Networks (DNNs)
#### Definition
When an ANN contains multiple hidden layers, it becomes a **Deep Neural Network (DNN)**.
#### Benefit
DNNs can learn highly complex and nonlinear relationships that traditional machine learning algorithms may struggle to identify.

### 4. How Deep Learning Works
Deep learning attempts to learn a mapping:
> Input (X) → Neural Network → Output (Y)

##### Example
Input:
* House size
* Bedrooms
* Location

Output:
* House Price

The network learns the relationship automatically.

### 5. Layers and Transformations
Each neuron performs calculations using:
#### Inputs
Feature values from previous layers.
#### Weights
Numeric values that determine feature importance.
#### Activation Functions
Transform values and help learn complex patterns.
#### Process
> Input → Multiply by Weights → Apply Function → Pass to Next Layer

Each layer learns increasingly sophisticated representations of data.

### 6. Deep Learning Training Process
The goal is to minimize prediction errors.
#### Step 1: Feed Data Forward
Input features enter the network.
> Input Layer → Hidden Layers → Output Layer

This process is called: **Forward Propagation**
#### Step 2: Generate Predictions
The model creates predicted values:
> `ŷ` (Predicted)

##### Example:
> Predicted Price = ₹10,00,000

#### Step 3: Calculate Loss
Compare:
> Actual Value (`y`) vs Predicted Value (`ŷ`)

##### Loss Function
Measures prediction error.

##### Example:
$$\text{Loss} = \text{Actual} - \text{Predicted}$$
The larger the difference, the larger the loss.

#### Step 4: Backpropagation
The network calculates how much each weight contributed to the error.
##### Purpose
Determine which weights need adjustment.

This process is called: **Backward Propagation (Backpropagation)**

#### Step 5: Update Weights
Weights are adjusted to reduce future errors.

Most commonly using: **Gradient Descent**
##### Goal
Move weights in the direction that minimizes loss.

#### Step 6: Repeat
The process repeats many times until the loss becomes very small.

These repetitions are called: **Epochs**

### 7. Deep Learning Workflow
> Training Data → Forward Propagation → Prediction → Loss Calculation → Backpropagation → Weight Updates → Repeat (Epochs)

### 8. Seven-Step Neural Network Process
According to the lesson:
#### Step 1
Define training and validation datasets.
#### Step 2
Features enter the input layer and move through neurons.
#### Step 3
Output layer generates predictions.
#### Step 4
Loss function compares:
> Actual (`y`) vs Predicted (`ŷ`)

#### Step 5
Optimization function calculates impact of each weight.
#### Step 6
Updated weights are backpropagated through the network.
#### Step 7
Repeat for multiple epochs until loss is minimized.

### 9. Key Deep Learning Algorithms and Concepts
#### Gradient Descent
##### Purpose
Optimization algorithm used to minimize prediction errors.
##### Goal
Find the set of weights that produces the lowest loss.

#### Loss Function
##### Purpose
Measures how wrong a prediction is.
> Lower Loss = Better Model

#### Forward Propagation
##### Purpose
Calculate predictions.
##### Direction
> Input → Output

#### Backpropagation
##### Purpose
Update model weights based on errors.
##### Direction
> Output → Input
