8. Time Travelling with Delta Tables
Explanation

Time Travel allows querying or restoring previous versions.

Query by Timestamp
SELECT * FROM employees
TIMESTAMP AS OF '2024-09-15T11:59:18Z'

Query by Version
SELECT * FROM employees
VERSION AS OF 2

PySpark Version
.option("versionAsOf",2)

Restore a Previous Version
RESTORE employees
TO VERSION AS OF 2

Important Learning

RESTORE does not overwrite history.

Instead:

Creates a new version
Copies older state
Preserves audit history
Summary

Time Travel is one of Delta Lake's most powerful features and a favorite DP-600 exam topic.

9. Managed & External Delta Tables
Explanation

This is one of the most important DP-600 concepts.

Managed Delta Table
Characteristics
Data stored in Tables folder
Fabric manages metadata
Fabric manages data lifecycle
Dropping table removes:
Metadata
Data files
External Delta Table
Characteristics
Data stored outside Tables area
User controls storage location
Fabric manages only metadata
Dropping table removes:
Metadata only
Data files remain intact
Comparison Table
Feature	Managed Table	External TableData Location	Tables folder	Custom location
Metadata Managed	Yes	Yes
Data Managed	Yes	No
Drop Table Deletes Data	Yes	No
Fabric Controls Storage	Yes	No
Summary

Managed tables are simpler, while external tables provide flexibility and independent data lifecycle management.

10. Creating Managed Delta Tables from Spark
Explanation

Managed Delta Table creation:

df.write.format("delta") \
.saveAsTable("phone_sales")

Outcome
Table appears under Tables section.
Data stored within Fabric-managed location.
Delta metadata managed automatically.
Summary

Using saveAsTable() without specifying a path creates a managed Delta Table.

11. Creating External Delta Tables from Spark
Explanation

External table creation:

df.write.format("delta") \
.saveAsTable(
"phone_sales_external",
path="Files/external")

Outcome
Metadata registered as table
Data stored outside managed Tables area
Fabric tracks metadata only
Another Scenario

Saving only Delta files:

df.write.format("delta") \
.save("Files/delta/phone_sales")


Result:

Delta format files created
No table registration
Not queryable via SQL endpoint
Summary

External tables separate metadata management from data storage, while Delta files alone are simply storage artifacts.
