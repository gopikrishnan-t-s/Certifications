## **2. Semantic Models in Fabric**
### Explanation
A Semantic Model is a structured business-friendly representation of data that sits between Fabric storage and Power BI reports.
### Purpose
Instead of Power BI accessing raw warehouse or lakehouse data directly, the Semantic Model:
* Organizes data
* Defines relationships
* Optimizes performance
* Provides reusable business calculations
### Core Components
#### Storage Modes
Controls:
* Data access
* Data freshness
* Performance
#### Relationships
Defines:
* Table connections
* Cardinality
* Filter propagation
#### Additional Components
* DAX expressions
* Measures
* Hierarchies
* Calculated columns
### Summary
Semantic Models transform raw Fabric data into analytics-ready structures for Power BI and reporting.

## **3. Working with Semantic Models**
### Explanation
Semantic Models are built using DAX (Data Analysis Expressions).
### DAX Uses
#### Measures
Example:
```dax
Average Call Duration =
AVERAGE(CallRecords[CallDuration])
```
#### Calculated Columns
Create derived fields.
#### Calculated Tables
Create additional modeling structures.
### XMLA Endpoint
Enables external connectivity.

Supported Tools:
* SSMS
* DAX Studio
* Tabular Editor
* Excel
### Uses
* Model Management
* Deployment
* Performance Tuning
* Monitoring
### Summary
DAX powers calculations, while XMLA allows enterprise management of Semantic Models.

## **4. Import and DirectQuery Storage Modes**
### Import Mode
#### How it Works
Data is imported into Power BI memory (VertiPaq).
#### Advantages
✅ Very fast performance \
✅ Complex calculations supported \
✅ Offline analysis possible
#### Limitations
❌ Requires scheduled refreshes \
❌ Data may become stale \
❌ Memory limitations
### DirectQuery Mode
#### How it Works
Queries execute directly against the source.
#### Advantages
✅ Real-time data \
✅ Supports massive datasets \
✅ Reduced memory usage
#### Limitations
❌ Slower performance \
❌ Depends on source performance \
❌ Limited modeling capabilities
### Summary
| Storage Mode | Performance | Freshness |
| --- | --- | --- |
| Import | High | Lower |
| DirectQuery | Lower | Real-Time |

## **5. Direct Lake and Composite Storage Modes**
### Direct Lake Mode
#### Most Important DP-600 Topic
Direct Lake combines:
* Import Performance
* DirectQuery Freshness
#### Key Features
* Reads directly from OneLake
* Uses Delta Tables
* Loads columns on demand
* Supports huge datasets
#### Benefits
✅ Near real-time access \
✅ In-memory performance \
✅ No full data import
#### Fallback Scenario
Direct Lake falls back to DirectQuery if:
* Dataset exceeds memory limits
* SKU thresholds exceeded
* Capacity limits exceeded
### Composite Mode
Combines:
* Import Tables
* DirectQuery Tables

Example:
| Table | Storage Mode |
| --- | --- |
| Historical Data | Import |
| Live Transactions | DirectQuery |
### Summary
Direct Lake is the preferred Fabric storage mode and a favorite DP-600 exam topic.

## **6. Table Relationships in Semantic Models**
### Explanation
Relationships allow data from multiple tables to work together.
### Cardinality Types
#### One-to-One
Employee ↔ Employee Profile
#### One-to-Many
Customer → Orders
#### Many-to-One
Orders → Customer
#### Many-to-Many
Students ↔ Courses
### Relationship Settings
#### Active Relationship
Used by default in reports.
#### Referential Integrity
Improves performance when guaranteed.
#### Cross Filter Direction
##### Single Direction (Recommended)
> Dimension → Fact

##### Both Direction
Used carefully for many-to-many scenarios.
### Summary
Relationships define how data connects and flows inside Power BI reports.

## **7. The Star Schema**
### Explanation
The preferred analytical model in Fabric.
### Structure
#### Fact Table
Stores metrics.

Examples:
* Sales
* Revenue
* Profit
* Quantity
#### Dimension Tables
Store descriptive information.

Examples:
* Customer
* Product
* Date
* Promotion
### Relationship Rule
> Dimension (1) → Fact (*)

### Benefits
✅ Faster queries \
✅ Simpler reporting \
✅ Better Power BI performance
### Summary
Star Schema is the recommended DP-600 modeling pattern.

## **8. Normalized and Denormalized Data**
### Normalized Data
#### Characteristics
* Data split into many tables
* Reduced redundancy
* Better integrity
#### Advantages
✅ Consistent data \
✅ Less duplication
#### Disadvantages
❌ More joins \
❌ Slower analytics
### Denormalized Data
#### Characteristics
* Data combined into fewer tables
* Redundant attributes allowed
#### Advantages
✅ Faster analytics \
✅ Simpler queries
#### Disadvantages
❌ More storage \
❌ Possible duplication
### Summary
Analytics workloads generally favor denormalized structures.

## **9. Star Schema vs Snowflake Schema**
### Star Schema
#### Characteristics
* Denormalized Dimensions
* Fewer joins
* Better query performance

✅ Reporting Friendly
#### Snowflake Schema
##### Characteristics
* Normalized Dimensions

* Less redundancy

✅ Better storage efficiency

## **12. Slowly Changing Dimensions (SCDs)**
### Explanation
Used to track changes in dimension data over time.
### Type 1
Overwrite existing data.

✅ Simple \
❌ No History

### Type 2
Add new record.

✅ Full History \
❌ More Storage

### Type 3
Add Previous Value Column.

✅ Limited History \
❌ Tracks only recent changes

### Type 4
Separate History Table.
### Type 5
Combination of Type 1 + Type 4.
### Type 6
Combination of Types 1, 2, and 3.
### DP-600 Focus
Most Important:
* Type 1
* Type 2
* Type 3

## **13. Creating Hierarchies in Models**
### Explanation
Hierarchies organize fields for drill-down analysis.
### Example 1
```
Country
  └─ State
       └─ City
```
### Example 2
```
Category
   └─ Subcategory
```
### Benefits
✅ Logical navigation \
✅ Drill down capability \
✅ Better reporting experience
### Important Direct Lake Note
Unlike Import Mode:
* Direct Lake does NOT automatically generate date hierarchies.

Hierarchies must be created manually.
### Summary
Hierarchies improve data exploration and interactive analysis in Power BI.
