## **3. Dataflow Gen2**
### Explanation
Dataflow Gen2 is a **low-code/no-code** data transformation tool.

Designed for users who:
* Do not know SQL
* Do not know programming
* Need visual data transformation
### Features
#### Power Query Integration
Visual transformation interface.
#### M Language
Generated automatically behind the scenes.
#### 300+ Built-in Transformations
Examples:
* Filtering
* Joining
* Aggregation
* Data Cleaning
#### Single Destination Model
Output goes to one destination:
* Lakehouse
* Warehouse
* Azure SQL Database
### DP-600 Scenario
User wants transformation without coding → Dataflow Gen2
### Summary
Dataflow Gen2 is Fabric's primary low-code transformation tool based on Power Query and M language.

## **13. Staging in Dataflow Gen2**
### Explanation
Staging creates an intermediate storage layer.
### Why Use Staging?
#### Unsupported Transformations
Apply transformations that Fast Copy cannot handle.
#### Unsupported Destinations
Example: 

`Warehouse` 

requires staging.
#### Large Datasets
Improve execution and query folding.
#### Trade-Off
✅ Flexibility \
❌ Additional processing overhead
### Summary
Staging is used when Fast Copy limitations require intermediate processing before final loading.

## **15. Incremental Refresh**
### Explanation
Incremental Refresh updates only new or changed data.
### Requirements
Must have:
```
DateTime Column
```
### How It Works
1. Data partitioned into buckets.
2. Changed buckets detected.
3. Only refreshed partitions processed.
### Benefits
✅ Faster refreshes \
✅ Lower resource usage \
✅ Better scalability \
✅ Near real-time updates
### Supported Destinations
* Azure SQL Database
* Fabric Warehouse
* Azure Synapse Analytics
### Summary
Incremental Refresh avoids full refreshes by processing only changed data.
