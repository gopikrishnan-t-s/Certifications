## **2. Structured, Unstructured, and Semi-structured Data**
### Explanation
The course begins by defining the three major categories of data:
#### Structured Data
* Organized into rows and columns.
* Uses predefined schemas.
* Examples: SQL tables, spreadsheets.
#### Semi-Structured Data
* Contains organizational markers but no rigid schema.
* Examples: JSON, XML, HTML, YAML.
#### Unstructured Data
* No predefined format.
* Examples:
    * Images
    * Videos
    * Documents
    * Social media content

The video also discusses Big Data concepts:
* Volume
* Velocity
* Variety

and how modern analytics, AI, and cloud technologies help organizations extract value from data.

### Summary
Understanding data types is essential because storage, processing, and analytics approaches vary depending on whether the data is structured, semi-structured, or unstructured.

## **3. Data Warehouses for Structured Data**
### Explanation
A Data Warehouse is a centralized repository designed primarily for structured data and analytical reporting.
### Key Characteristics
* Schema-on-write
* ETL-based loading
* High query performance
* Historical data storage
* BI and reporting-focused
### Popular Warehouse Solutions
* Amazon Redshift
* Google BigQuery
* Snowflake
* Azure Synapse Analytics
* Oracle Exadata
### Benefits
* Fast analytics
* Strong governance
* Optimized reporting
* Centralized enterprise data
### Limitations
* Expensive
* Rigid schemas
* Difficult to work with unstructured data
### Summary
Data warehouses provide high-performance analytics for structured data but lack flexibility for modern data types.

## **4. Data Lakes vs Data Warehouses**
### Explanation
The course compares two major data storage approaches.
#### Data Warehouse
* Structured data
* Schema-on-write
* ETL processing
* Business analytics focused
#### Data Lake
* Structured, semi-structured, and unstructured data
* Schema-on-read
* ELT processing
* Machine learning and big-data focused
##### Popular Data Lakes
* Azure Data Lake Storage (ADLS)
* Amazon S3
* Google Cloud Storage
* IBM Cloud Object Storage
### Summary
Data lakes offer flexibility and low-cost storage, while data warehouses provide optimized querying and reporting.

## **5. Data Lakehouses**
### Explanation
A Data Lakehouse combines the benefits of both:
* Data Lakes
* Data Warehouses
### Fabric Lakehouse Structure
#### Files Section
Acts like a data lake:
* Raw files
* Semi-structured data
* Unstructured data
#### Tables Section
Acts like a data warehouse:
* Delta tables
* Structured analytics-ready data
### Benefits
* Unified architecture
* Lower storage costs
* Support for Spark and BI
* Supports ML and reporting in one environment
### Summary
Lakehouses eliminate the separation between business analytics and data engineering workloads.

## **12. Delta Lakes and Delta Tables**
### Explanation
Delta Lake is the storage technology that powers Lakehouses.
### Features
* ACID Transactions
* Time Travel
* Schema Enforcement
* Data Versioning
* Batch & Streaming Support
### Architecture
* Parquet files store the data.
* Delta Log stores transaction metadata.
### Summary
Delta Lake bridges the gap between data lakes and data warehouses by adding reliability and version control.
