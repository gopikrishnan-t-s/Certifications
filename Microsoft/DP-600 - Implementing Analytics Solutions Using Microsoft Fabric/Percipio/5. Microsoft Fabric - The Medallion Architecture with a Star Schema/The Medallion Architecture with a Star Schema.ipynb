## **2. The Medallion Architecture**
### Explanation
The Medallion Architecture is a layered approach that progressively improves data quality.
### Bronze Layer (Raw Data)
#### Purpose:
* Landing zone for source data
* Preserve original data
* Support auditing and recovery
#### Characteristics:
* Raw data
* JSON, CSV, XML, logs
* Stored in Files section or external storage
* No transformations applied
### Silver Layer (Validated Data)
#### Purpose:
* Clean and standardize data
#### Operations:
* Remove duplicates
* Handle nulls
* Apply business rules
* Data enrichment
* Data validation
* Referential integrity checks
### Gold Layer (Business Layer)
#### Purpose:
* Analytics and reporting
#### Operations:
* Aggregations
* Summaries
* Star Schema implementation
* Reporting optimization
* Semantic models creation
### Summary
| Layer | Purpose | Storage |
| --- | --- | --- |
| Bronze | Raw Data | Files / Shortcuts |
| Silver | Cleaned Data | Delta Tables |
| Gold | Business Data | Delta Tables / Warehouse |

## **3. The Star Schema in Semantic Models**
### Explanation
The Gold Layer commonly uses a Star Schema.
### Core Components
#### Fact Table
##### Stores:
* Measures
* Metrics
* Aggregations
##### Examples:
* Revenue
* Quantity
* Profit
* Sales
##### Foreign Keys
Examples:
* CustomerID
* ProductID
* OrderID
* OrderDate
### Dimension Tables
Store descriptive information.
#### Examples:
##### Customer Dimension
* CustomerID
* Name
* City
* Country
##### Product Dimension
* ProductID
* Category
* Segment
##### Date Dimension
* Date
* Month
* Year
##### Order Dimension
* Order Priority
* Ship Mode
#### Relationship Type
> Dimension → Fact

Relationship: \
**One-to-Many**
> Dimension Table (1) → Fact Table (*)

##### Benefits
* Faster queries
* Simplified reporting
* Better Power BI performance
* Easy business understanding
### Summary
The Star Schema is the preferred Gold Layer design because it separates business measurements from descriptive information and simplifies analytics.

## **4. The Bronze Layer in the Medallion Architecture**
### Explanation
The Bronze Layer is implemented in the Files section of a Lakehouse.
### Key Concepts
#### Incremental Processing
Data arrives in multiple files:
```
Part1.csv
Part2.csv
Part3.csv
```
Instead of loading each file separately:
```python
load("Files/bronze/*.csv")
```
All files are processed automatically.
#### Data Cleaning Activities
Performed before Silver Layer loading:
* Remove duplicate rows
* Remove null values
* Remove outliers
* Standardize formats
* Create calculated columns

Example:

$\text{Profit Margin} = \frac{\text{Profit}}{\text{Sales}}$

### Summary
The Bronze Layer preserves raw source files while preparing them for Silver Layer transformation.

### Merge and Upsert
Key DP-600 Topic

## **9. Partitioning and Table Maintenance**
### Explanation
Large Delta Tables require optimization.
### Partitioning
Partition data by a frequently filtered column.
#### Example:
```python
.partitionBy("State")
```
#### Benefits
Without partitioning:
> Query scans all files

With partitioning:
> Query scans relevant partition only

### Optimize
#### Purpose:
* Combine many small Parquet files into larger files.
#### Benefits:
* Faster reads
* Better compression
* Better performance
### V-Order
Microsoft Fabric-specific optimization.
#### Provides:
* Better sorting
* Better compression
* Faster query execution
### Vacuum
#### Purpose:
* Remove old unused files.
#### Default retention:
> 7 Days

#### Benefits:
* Lower storage costs
* Cleaner Delta Tables
#### Important:
Reducing retention impacts Time Travel availability.
### Summary
Partitioning, Optimize, V-Order, and Vacuum are critical maintenance techniques for high-performance Delta Tables.
