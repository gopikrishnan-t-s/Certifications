## **3. DAX Context (Very Important for DP-600)**
### Row Context
Calculation performed row-by-row.
#### Example:
```dax
Profit = Sales - Cost
```
Computed separately for every row.
### Filter Context
Active report filters affect calculations.
#### Example:
```
Country = India
```
Then:
```dax
Total Sales
```
returns only India's sales.
### Query Context
Affected by report structure.
#### Example:
```
Sales by Region
```
DAX calculates separately for each region shown in the visual.
### DP-600 Exam Focus
| Context Type | Definition |
| --- | --- |
| Row Context | Current row |
| Filter Context | Current filters |
| Query Context | Current visualization |

## **6. Measures vs Calculated Columns**
### Measures
#### Characteristics
* Dynamic
* Evaluated at query time
* Filter-aware
* Not physically stored
#### Example:
```dax
Total Sales =
SUM(Sales[Amount])
```
#### Uses
✅ Totals \
✅ KPIs \
✅ Ratios \
✅ Percentages \
✅ Time Intelligence
### Calculated Columns
#### Characteristics
* Row-by-row calculations
* Stored physically
* Not filter-sensitive
#### Example:
```dax
Full Name =
FirstName & " " & LastName
```
#### Uses
✅ Categorization \
✅ Derived Fields \
✅ Keys \
✅ Labels
### Exam Comparison
| Feature | Measure | Calculated Column |
| --- | --- | --- |
| Dynamic | ✅ | ❌ |
| Stored | ❌ | ✅ |
| Filter Sensitive | ✅ | ❌ |
| Memory Usage | Low | High |

## **8. DAX Queries**
### Definition
DAX Queries retrieve and summarize data.

Usually begin with:
```dax
EVALUATE
```
### `SELECTCOLUMNS`
Used to retrieve selected columns.
```dax
EVALUATE
SELECTCOLUMNS(...)
```
Returns all rows.
### `SUMMARIZECOLUMNS`
Used for:
* Grouping
* Aggregation
* Summarization
```
EVALUATE
SUMMARIZECOLUMNS(...)
```
Returns grouped data.
### Exam Rule
* `SELECTCOLUMNS` = Return rows
* `SUMMARIZECOLUMNS` = Group rows

## **11. Filtering in DAX**
### `FILTER()`
Returns rows meeting conditions.
#### Example:
```dax
FILTER(
   Customers,
   Customers[CreditScore] >= 700
)
```
### `TREATAS()`
Applies filter values.
#### Example:
```dax
TREATAS(
 {"Married"},
 Customers[Marital_Status]
)
```
### `KEEPFILTERS()`
Adds filters without replacing existing filters.

## **12. CALCULATE Function (Most Important Topic)**
### Definition
`CALCULATE` modifies filter context.
### Syntax:
```dax
CALCULATE(
   Expression,
   Filter
)
```
#### Example
```dax
SelfEmployed_Balance =
CALCULATE(
 SUM(Customer[AccountBalance]),
 Customer[EmploymentStatus]
     = "Self-Employed"
)
```
### What CALCULATE Does
1. Evaluates expression.
2. Modifies filter context.
3. Applies new filters.
### DP-600 Priority
⭐⭐⭐⭐⭐ CALCULATE

## **13. CALCULATE + FILTER**
### Example:
```dax
CALCULATE(
 SUM(AccountBalance),
 FILTER(
   Customers,
   Customers[MaritalStatus]="Married"
   &&
   Customers[AccountType]="Checking"
 )
)
```
### Use Case
Complex multi-column filtering.

## **14. ALL Function**
### Definition
Removes filters.
#### Example:
```dax
CALCULATE(
 SUM(AccountBalance),
 ALL(Customers)
)
```
### Result
Always returns:
```
Grand Total
```
ignoring report filters.
### Example
| Account Type | Grand Total |
| --- | --- |
| Business | Same value |
| Checking | Same value |
| Savingsn | Same value |

Because ALL removes filtering.
