## **3. The T-SQL Endpoint & Semantic Models**
### Explanation

Provides SQL-based access to Lakehouse tables.
##### Characteristics
* Read-only for Lakehouses
* Used for querying Delta tables
* Supports Visual Query Editor
* Works only with table data
##### Spark vs T-SQL Endpoint
| Feature | Spark | T-SQL Endpoint |

| Query Data | ✅ | ✅ |

Used to:
* Define relationships
* Create measures
* Build Star Schema
* Power Power BI reports

### Summary
T-SQL is mainly for querying Lakehouse tables, while semantic models organize business data for analytics and reporting.

## **4. Data Lakehouse Shortcuts**

* ADLS Gen2
* Amazon S3
* Google Cloud Storage
### Benefits
* No duplicated storage
* Dynamic updates
* Single source of truth
* Simplified access to external data
### Summary

## **6. Creating Dynamic Shortcuts in Fabric**
### Explanation
The course demonstrates how shortcut data updates automatically when the source data changes.
### Key Findings
* Shortcut files update dynamically.
* Delta Tables do not automatically update.
* Managed Delta Tables require refresh/reload actions.
### Exam Tip
A shortcut is dynamic. A managed Delta table is not automatically synchronized.
### Summary
Changes in source storage immediately appear through shortcuts, but managed Delta tables must be refreshed manually.

## **12. The Fabric Capacity Metrics App**
### Explanation
Introduces Fabric capacity management.
### Key Terms
#### Capacity
Pool of compute, storage, and memory resources.
#### Capacity Units (CUs)
Measurement of compute resources.
#### SKU
Capacity size designation (F2, F4, F64, etc.).
### Overages and Throttling
#### Overage Protection
First 10 minutes of excess usage.
#### Interactive Delay
10-60 minutes over limit.
#### Interactive Rejection
More than 60 minutes over limit.
#### Background Rejection
More than 24 hours over limit.
### Summary
Capacity Metrics App helps monitor resource usage, optimize performance, and prevent throttling.
