## **1. Course Overview**
### Explanation
Data Pipelines are the second major Data Factory component (alongside Dataflow Gen2). They are used to orchestrate, automate, schedule, and monitor data workflows by connecting multiple activities together into an end-to-end ETL process.
### Summary
* Data Pipelines
* Activities
* Copy Activity
* Script Activity
* Stored Procedures
* Notebooks
* Spark Jobs
* Data Flow Activity
* Monitoring Hub
* Email Notifications
* Pipeline Triggers

## **2. Data Pipelines vs Dataflows**
### Data Pipelines
Used for:

✅ Orchestration \
✅ Scheduling \
✅ Workflow Automation \
✅ Running Activities \
✅ Control Flow Logic

Examples:
* Copy data
* Run Notebook
* Execute SQL Script
* Run Stored Procedure
* Send Email
### Dataflow Gen2
Used for:

✅ Data Transformation \
✅ Data Cleansing \
✅ Low-Code ETL

Uses:
* Power Query
* M Language

## **3. Data Pipelines**

Individual tasks.

* Data Flow
#### Control Flow
Supports:
* If-Else
* ForEach
* Do Until
* Branching
#### Triggers
Supports:
* Manual
* Scheduled
* Event-Based
### Summary
Pipelines orchestrate and automate data processing using activities and control-flow logic.

## **5. Control Flow and Triggers**
### Explanation
Pipelines support advanced workflow control.
### If Condition
```
IF success
   Continue
ELSE
   Alternate path
```
### ForEach
Loops through collections.
### Do Until
Repeats until condition becomes true.
### Triggers
#### On-Demand
Manual execution.
#### Scheduled
Time-based execution.
#### Event-Based
Triggered by:
* File arrival
* Data change
* External event
### Summary
Control-flow activities make pipelines dynamic and intelligent.

## **6. Copy Activity (Very Important)**
### Purpose
Moves data between sources and destinations.
### Common Sources
* ADLS Gen2
* SQL Server
* PostgreSQL
* Snowflake
* Azure SQL
* Fabric Lakehouse
* Fabric Warehouse
### Common Destinations
* Lakehouse
* Warehouse
* Azure SQL
* Blob Storage
### Supported Operations
#### Serialization
Convert formats.
#### Compression
Compress/decompress.
#### Schema Mapping
Map source columns to destination.
#### Staging
Use interim storage for large transfers.
#### Parallelism
Improve performance.
### DP-600 Exam Tip
> Fast data movement → Copy Activity

### Summary
Copy Activity is Fabric's primary data movement component.

## **7. Script Activity**
### Purpose
Executes SQL scripts.
### Supported Platforms
✅ Fabric Warehouse \
✅ Azure SQL Database \
✅ SQL Server \
✅ Snowflake \
✅ Synapse Analytics
### NOT Supported:
❌ Lakehouse \
❌ KQL Database
### Types
#### Query Script
Returns results.
```sql
SELECT *
FROM Sales
```
#### Non-Query Script
Does not return results.
```sql
CREATE TABLE
UPDATE
DELETE
```
### Summary
Use Script Activity when executing SQL code inside a pipeline.

> Pipeline + Dataflow

### Summary

## **13. Activity Outcomes (Important Exam Topic)**
Each activity exposes four connectors.
| Connector  | Executes When                          |
| ---------- | -------------------------------------- |
| Skip       | Activity Skipped                       |
| Success ✅  | Activity Succeeded                     |
| Failure ❌  | Activity Failed                        |
| Complete ➜ | Activity Finished (Success or Failure) |

### Example
```
Notebook
    ↓ Success
Dataflow
    ↓ Success
Email Success
```
```
Dataflow
    ↓ Failure
Email Failure
```
### DP-600 Tip
**Success vs Failure vs Complete connectors** are frequently tested.
