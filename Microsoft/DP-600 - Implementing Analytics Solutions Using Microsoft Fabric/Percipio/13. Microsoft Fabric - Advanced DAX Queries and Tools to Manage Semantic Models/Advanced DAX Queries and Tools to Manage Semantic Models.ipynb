## **2. DAX Iterators (Very Important)**
### What Are Iterators?
Iterators perform calculations row-by-row and then aggregate results.
### Why Use Them?
Normal functions:
```dax
SUM()
AVERAGE()
```
operate on a single column.

Iterators allow:
```dax
Price × Quantity
Balance + Overdraft
```
using multiple columns per row.
### Common Iterators
| Function | Purpose |
| --- | --- |
| `SUMX` | Sum expression per row |
| `AVERAGEX` | Average expression per row |
| `MEDIANX` | Median expression per row |
| `MAXX` | Maximum value |
| `MINX` | Minimum value |
| `RANKX` | Ranking |
| `FILTER` | Row filtering iterator |
### Example
```dax
SUMX(
 Customer,
 Customer[Balance]
 + Customer[Overdraft]
)
```
Computes total withdrawal amount.
### DP-600 Tip
**Functions ending in X = Iterators**

Examples:
```
SUMX
AVERAGEX
MEDIANX
RANKX
```

## **3. DAX Variables (VAR)**
### Purpose
Store intermediate calculations.
### Benefits:
✅ Better performance \
✅ Better readability \
✅ Avoid repeated calculations
### Syntax
```dax
VAR FilteredSales =
    FILTER(...)

VAR TotalSales =
    SUMX(...)

RETURN
    TotalSales
```
### Benefits
| Without VAR | With VAR |
| --- | --- |
| Repeated calculations | Reused calculations |
| Hard to read | Clean |
| Slower | Faster |
### Exam Rule
**VAR + RETURN = Advanced DAX** Best Practice

## **6. Table Filtering Functions**
### `FILTER`
Returns subset of rows.
```dax
FILTER(
 Sales,
 Sales[Amount] > 1000
)
```
### `CALCULATETABLE`
Returns filtered table.
```dax
CALCULATETABLE(...)
```
### `ALL`
Removes filters.
```dax
ALL(Sales)
```
### `ALLEXCEPT`
Removes all filters except selected ones.
```dax
ALLEXCEPT()
```
### `TOPN`
Returns Top `N` records.
```dax
TOPN(5,...)
```

## **7. Window Functions (New DP-600 Favorite)**
### Purpose
Advanced analytics over ranges of rows.

Uses:

✅ Moving averages \
✅ Running totals \
✅ Rankings \
✅ Trends
### Window Functions
#### `WINDOW()`
Define row ranges.
#### `OFFSET()`
Relative rows.
#### `RANKX()`
Ranking.
### Relative Window
```dax
WINDOW(
 -2, REL,
 0, REL
)
```
Current row + previous 2 rows.
### Absolute Window
```dax
WINDOW(
 0, ABS,
 0, REL
)
```
Start of dataset to current row.
### Important Parameters
| Parameter | Purpose |
| --- | --- |
| `REL` | Relative row |
| `ABS` | Absolute row |
| `ORDERBY` | Sort rows |
| `PARTITIONBY` | Group windows |
| `KEEP` | Respect filters |
### Example Uses
✅ 3-Day Moving Average \
✅ Running Total \
✅ Rolling Minimum \
✅ Ranking

## **8. Calculation Groups**
### Most Important Productivity Feature
Used to create reusable calculations.
### Typical Use
Without Calculation Groups:
```dax
YTD_Sales
MTD_Sales
YTD_Profit
MTD_Profit
```
With Calculation Groups:
```dax
TOTALYTD(
 SELECTEDMEASURE()
)
```
### Benefits
✅ Less duplication \
✅ Easier maintenance \
✅ Consistent calculations
