## **Introduction to end-to-end analytics using Microsoft Fabric**

### **Introduction**

**Microsoft Fabric** is an end-to-end analytics platform that unifies the entire data lifecycle in a single environment. Instead of using separate tools for data ingestion, preparation, governance, analytics, and AI, organizations can perform all these tasks within Fabric.
#### Why Fabric Matters
* Reduces complexity caused by disconnected data tools and teams.
* Provides a single platform for data engineers, analysts, data scientists, and business users to collaborate.
* Eliminates duplicated effort by using the same governed data across analytics and AI workloads.
#### Core Foundation: OneLake
Fabric is built on **OneLake**, a unified data lake that serves as the central storage layer for organizational data. Because data is stored and managed in one place, it can be used consistently across all Fabric experiences.
#### Fabric Workloads
Fabric brings together capabilities for:
* Data ingestion
* Data engineering
* Data integration
* Data warehousing
* Data science
* Real-time analytics
* Business intelligence and reporting
#### Analytics and AI Together
A key benefit of Fabric is that the data prepared and governed for analytics is automatically available for AI scenarios, including:
* Copilot
* Data agents
* Fabric IQ
* Machine learning models
* Intelligent agents

This means governance, quality, and preparation work done for reporting and dashboards also supports AI initiatives, creating a single trusted data foundation.
#### In One Sentence
Microsoft Fabric is a unified analytics and AI platform built on OneLake that enables organizations to ingest, manage, govern, analyze, and use data for AI workloads from a single integrated environment.

### **Explore end-to-end analytics with Microsoft Fabric**

#### 1. Microsoft Fabric as a Unified SaaS Analytics Platform
Microsoft Fabric is a **software-as-a-service (SaaS)** analytics platform that brings together multiple analytics capabilities into a single integrated environment. Instead of managing separate products for data engineering, integration, warehousing, science, and BI, organizations can use one platform.
##### Benefits:
* Simplified analytics architecture
* Reduced operational complexity
* Improved scalability
* Lower costs
* Cloud-based accessibility from anywhere
#### 2. OneLake: The Foundation of Fabric
**OneLake** is Fabric's unified, centralized data lake.
##### Key characteristics:
* Built on **Azure Data Lake Storage Gen2 (ADLS Gen2)**.
* Stores data once and makes it available across all Fabric workloads.
* Supports open formats such as:
    * Delta
    * Parquet

    * JSON
* Eliminates the need to duplicate or move data between tools.
##### Why OneLake Matters
* Single source of truth for organizational data.
* All Fabric compute engines read and write directly to OneLake.
* Tabular data is stored in **Delta-Parquet** format.
* Enables seamless collaboration across teams and workloads.

### **Understand OneLake**

#### What is OneLake?
**OneLake** is the foundational storage layer of Microsoft Fabric.
##### Key characteristics:
* **Tenant-wide data lake** automatically available when Fabric is enabled.
* Provides **a single copy** of data for the entire organization.
* Shared by all Fabric workloads.
* Eliminates data silos and unnecessary data duplication.
* Acts as a single source of truth.
##### Why It Matters
In traditional systems, different teams often maintain separate copies of the same data. With OneLake:
* All workloads access the same underlying data.
* Updates are immediately available to everyone.
* Storage costs and data inconsistencies are reduced.
#### OneLake Catalog
The **OneLake Catalog** is a searchable inventory of data assets across the Fabric tenant. You can:
* Search by asset name.
* Browse by workspace.
* Browse by domain.
* View metadata.
* View asset owners.
* Track lineage.
##### Governance Features
OneLake integrates with Microsoft Purview to provide:
* Data classification
* Sensitivity labels
* Data lineage tracking
* Access control and security
#### Types of Data Stored in OneLake
##### 1. Tables
Stored in:
* Lakehouses
* Warehouses
* Eventhouses
##### 2. Files
Supported formats include:
* Parquet
* CSV
* JSON
* Other file formats
##### 3. Shortcuts
References to data stored elsewhere, such as:
* Azure Data Lake Storage
* Amazon S3
* Other OneLake locations

Benefits:
* No data duplication
* Data remains in its original location
* Easier governance and management
##### 4. Semantic Models
Used by:
* Power BI
* Analytics workloads
* AI workloads
#### Open Data Formats
OneLake uses **open formats**, meaning data is not locked into proprietary storage formats.
##### Default Table Format
* Delta Lake
##### Physical Storage Format
* Parquet files

Benefits:
* Interoperability
* Accessibility by multiple tools
* Flexibility for future use
#### How Data Gets Into OneLake
##### Mirroring
Continuously replicates data from:
* SQL Server
* Azure SQL Database
* Azure Cosmos DB
* Snowflake

Changes in the source are automatically reflected in OneLake.
##### Pipelines
Use Data Factory capabilities to:
* Move data
* Transform data
* Load data into OneLake
##### Dataflows
Use Power Query to:
* Connect to sources
* Transform data
* Load data into OneLake
##### Streaming
Uses eventstreams to ingest:
* IoT data
* Application logs
* Clickstream events
* Other real-time data
##### Direct Upload
Upload files directly through the Fabric interface.

OneLake is critical for AI experiences in Fabric.

### **Discover streaming data in Real-Time hub**

#### Why Real-Time Hub?
Unlike batch data stored in lakehouses and warehouses, streaming data arrives continuously:
##### Examples:
* Online purchases
* IoT sensor readings
* Application logs
* Inventory updates
* Clickstream events

The **Real-Time Hub** helps you discover and manage these streaming data sources across Fabric.

#### 1. What is Real-Time Hub?
**Real-Time Hub** is the centralized catalog for streaming data in Microsoft Fabric.
##### Comparison
| OneLake Catalog | Real-Time Hub |
| --- | --- |
| Batch data | Streaming data |
| Lakehouses | Eventstreams |
| Warehouses | KQL Tables |
| Semantic Models | Eventhouses |
##### Remember:
* OneLake Catalog = Stored data
* Real-Time Hub = Streaming data
#### 2. Key Components
##### Eventstreams
An **eventstream** is a continuous flow of real-time events.

Common sources:
* Azure Event Hubs
* Azure IoT Hub
* Apache Kafka
* Database CDC
* Custom applications

Examples:
* Customer purchases
* Sensor readings
* Web clicks
##### Eventhouses
An **eventhouse** is a container that stores streaming data.

It contains:
* One or more KQL databases
* Time-series and event-based data

Purpose:
* Organize and manage streaming analytics data.
##### KQL Databases and KQL Tables
KQL databases:
* Store streaming data.

KQL tables:
* Contain ingested events.

Benefits:
* Automatically indexed
* Optimized for high-speed analytics
* Support near real-time querying
##### Kusto Query Language (KQL)
KQL is used to:
* Query streaming data
* Analyze events
* Create real-time dashboards
* Generate alerts
#### 3. Discover Streaming Data

You can:
* Browse by workspace

## **Get started with lakehouses in Microsoft Fabric**

### **Introduction**

#### What is a Lakehouse?
A **Lakehouse** is a storage and analytics architecture that combines:
* **Data Lake capabilities** → store large volumes of structured, semi-structured, and unstructured data.
* **Data Warehouse capabilities** → query, analyze, and report on data efficiently.

In Microsoft Fabric, lakehouses are built on OneLake and use:
* **Apache Spark** for data engineering and large-scale processing.
* **SQL compute engines** for querying and analytics.
#### Why Organizations Need Lakehouses
Traditional data warehouses work well for structured data such as:
* Sales transactions
* Customer records
* Inventory data

However, modern organizations also generate:
* Application logs
* IoT sensor data
* Social media data
* JSON and CSV files
* External data feeds

Managing these different data types often requires separate systems, creating:
* Data silos
* Complex integrations
* Increased maintenance costs

A lakehouse solves this by providing a single platform for all data types.
#### Lakehouse Benefits
✅ Store structured, semi-structured, and unstructured data \
✅ Built on the centralized **OneLake** storage layer \
✅ Supports both **SQL** and **Spark** workloads \
✅ Reduces data movement and duplication \
✅ Scales to large data volumes \
✅ Supports analytics, reporting, machine learning, and AI

### **Describe lakehouse features and capabilities**

#### Why Lakehouses?
Traditional architectures force a choice:
| Data Lake                        | Data Warehouse                   |
| -------------------------------- | -------------------------------- |
| Flexible storage                 | Strong analytics                 |
| Handles any data type            | Structured data only             |
| Scalable                         | Optimized SQL querying           |
| Limited governance and analytics | Higher cost and less flexibility |

A **Lakehouse** combines the benefits of both.
##### Lakehouse = Data Lake + Data Warehouse
Benefits:
* Store all data types.
* Query with SQL.
* Process with Spark.
* Support analytics, BI, machine learning, and AI.
#### Lakehouse Structure
A Fabric Lakehouse contains two main areas:
##### 1. Tables Folder
Contains **Delta Lake tables**.

Features:
* Supports SQL queries via SQL Analytics Endpoint.
* Schema enforcement.
* ACID transactions.
* Accessible in Power BI.
* Automatic optimization and maintenance.

Use for:
* Reporting
* Analytics
* Semantic models
* BI workloads
##### 2. Files Folder
Contains raw files in their native format.

Supported formats:
* CSV
* JSON
* Parquet
* Images
* Documents
* Any file type

Features:
* Flexible storage.
* Data staging area.
* No schema enforcement.
* Cannot be queried directly with SQL.

Use for:
* Raw data ingestion
* Historical storage
* Pre-transformation staging
##### Common Workflow
```
Raw Files → Spark / Dataflows → Transform Data → Delta Tables → SQL / Power BI / AI
```
#### Delta Lake Tables
Delta Lake is the technology that powers lakehouse tables.
##### 1. ACID Transactions
Ensures:
* Consistency
* Reliability
* Concurrent reads and writes

Meaning multiple users can work with data safely.
##### 2. Schema Enforcement
Delta validates incoming data.

Benefits:
* Prevents invalid data loads.
* Maintains data quality.
##### 3. Time Travel
Delta keeps a transaction log.

Allows you to:
* Query previous versions.
* Recover from mistakes.
* Roll back changes.
##### 4. Efficient Updates and Deletes
Unlike traditional files:
* `UPDATE` supported
* `DELETE` supported
* `MERGE` supported

Without rewriting entire datasets.
##### Delta Architecture
Each Delta table consists of:
```
Parquet Data Files + Transaction Log
```
The transaction log records all changes and enables advanced capabilities.

Use SQL to query tables
When you create a lakehouse and define tables in it, a SQL endpoint is automatically created through which the tables can be queried using SQL SELECT statements.

At the top-right of the Lakehouse page, switch from Lakehouse to SQL analytics endpoint. Then wait a short time until the SQL analytics endpoint for your lakehouse opens in a visual interface from which you can query its tables.

Use the New SQL query button to open a new query editor, and enter the following SQL query:

sql
SELECT Item, SUM(Quantity * UnitPrice) AS Revenue
FROM sales
GROUP BY Item
ORDER BY Revenue DESC;
Note: If you are in a lab VM and have any problems entering the SQL query, you can download the 01-Snippets.txt file from https://github.com/MicrosoftLearning/mslearn-fabric/raw/main/Allfiles/Labs/01/Assets/01-Snippets.txt, saving it on the VM. You can then copy the query from the text file.

Use the ▷ Run button to run the query and view the results, which should show the total revenue for each product.
