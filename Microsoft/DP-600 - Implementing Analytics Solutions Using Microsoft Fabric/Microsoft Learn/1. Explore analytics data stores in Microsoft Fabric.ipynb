## **Introduction to end-to-end analytics using Microsoft Fabric**

### **Introduction**

**Microsoft Fabric** is an end-to-end analytics platform that unifies the entire data lifecycle in a single environment. Instead of using separate tools for data ingestion, preparation, governance, analytics, and AI, organizations can perform all these tasks within Fabric.
#### Why Fabric Matters
* Reduces complexity caused by disconnected data tools and teams.
* Provides a single platform for data engineers, analysts, data scientists, and business users to collaborate.
* Eliminates duplicated effort by using the same governed data across analytics and AI workloads.
#### Core Foundation: OneLake
Fabric is built on **OneLake**, a unified data lake that serves as the central storage layer for organizational data. Because data is stored and managed in one place, it can be used consistently across all Fabric experiences.
#### Fabric Workloads
Fabric brings together capabilities for:
* Data ingestion
* Data engineering
* Data integration
* Data warehousing
* Data science
* Real-time analytics
* Business intelligence and reporting
#### Analytics and AI Together
A key benefit of Fabric is that the data prepared and governed for analytics is automatically available for AI scenarios, including:
* Copilot
* Data agents
* Fabric IQ
* Machine learning models
* Intelligent agents

This means governance, quality, and preparation work done for reporting and dashboards also supports AI initiatives, creating a single trusted data foundation.
#### In One Sentence
Microsoft Fabric is a unified analytics and AI platform built on OneLake that enables organizations to ingest, manage, govern, analyze, and use data for AI workloads from a single integrated environment.

### **Explore end-to-end analytics with Microsoft Fabric**

#### 1. Microsoft Fabric as a Unified SaaS Analytics Platform
Microsoft Fabric is a **software-as-a-service (SaaS)** analytics platform that brings together multiple analytics capabilities into a single integrated environment. Instead of managing separate products for data engineering, integration, warehousing, science, and BI, organizations can use one platform.
##### Benefits:
* Simplified analytics architecture
* Reduced operational complexity
* Improved scalability
* Lower costs
* Cloud-based accessibility from anywhere
#### 2. OneLake: The Foundation of Fabric
**OneLake** is Fabric's unified, centralized data lake.
##### Key characteristics:
* Built on **Azure Data Lake Storage Gen2 (ADLS Gen2)**.
* Stores data once and makes it available across all Fabric workloads.
* Supports open formats such as:
    * Delta
    * Parquet
    * CSV
    * JSON
* Eliminates the need to duplicate or move data between tools.
##### Why OneLake Matters
* Single source of truth for organizational data.
* All Fabric compute engines read and write directly to OneLake.
* Tabular data is stored in **Delta-Parquet** format.
* Enables seamless collaboration across teams and workloads.
#### 3. OneLake Shortcuts
**Shortcuts** are references to data stored elsewhere.
##### They can point to:
* OneLake locations
* Azure Data Lake Storage
* Amazon S3
* Microsoft Dataverse
##### Advantages:
* Access data without copying it.
* Maintain consistency with the original source.
* Reduce storage costs and duplication.
* Keep Fabric synchronized with source systems.
#### 4. AI Readiness Through OneLake
Because all Fabric workloads store governed data in OneLake using open formats:
* Copilot can access the same data used for analytics.
* Data agents can use governed enterprise data.
* AI models and Fabric IQ can work with trusted datasets.
* No separate AI-specific data preparation pipelines are required.
##### Important concept:
The data governance and preparation work completed for analytics automatically supports AI workloads.
#### 5. Workspaces
Workspaces are logical containers used to organize Fabric assets.
##### A workspace can contain:
* Data
* Reports
* Notebooks
* Warehouses
* Pipelines
* Other Fabric items
##### Workspace Benefits
* Resource organization
* Team collaboration
* Access control
* Security isolation
##### Additional Capabilities
* Workspace-specific permissions
* Compute resource management
* Git integration for:
    * Version control
    * Change tracking
    * Team collaboration
    * Code history management
#### 6. Administration and Governance
Fabric governance is centralized through the **Admin Portal**.
##### Administrators can:
* Manage users and groups
* Configure permissions
* Manage data sources and gateways
* Monitor usage and performance
* Access admin APIs and SDKs
* Automate administrative tasks
#### 7. OneLake Catalog
The **OneLake Catalog** helps organizations govern and manage data.
##### It provides visibility into:
* Sensitivity labels
* Item metadata
* Refresh status
* Governance recommendations
##### This allows organizations to:
* Discover data easily
* Monitor governance compliance
* Improve data quality and security
#### Exam-Focused Points to Remember
* Fabric = Unified SaaS analytics platform 
* OneLake = Single logical data lake built on ADLS Gen2 
* Supported formats = Delta, Parquet, CSV, JSON 
* Shortcuts = Access external data without copying it
* Workspaces = Organization, collaboration, permissions, Git integration 
* Admin Portal = Centralized administration and governance 
* OneLake Catalog = Data discovery and governance insights 
* Governed OneLake data powers both analytics and AI (Copilot, Data Agents, Fabric IQ)

### **Explore data teams and Microsoft Fabric**

#### Microsoft Fabric and Team Collaboration
Microsoft Fabric improves collaboration by bringing all data and analytics capabilities into a single SaaS platform, eliminating data silos and reducing the need for multiple disconnected tools.
#### Traditional Challenges
In traditional analytics environments:
* **Data engineers** prepare data for analysts, requiring significant coordination.
* **Data analysts** often perform additional transformations before building reports, increasing effort and reducing efficiency.
* **Data scientists** struggle to integrate machine learning workflows with existing analytics systems.
* Multiple handoffs between teams can cause delays, duplicated work, and misunderstandings.
#### How Fabric Improves Collaboration
Fabric enables different data roles to work from the same data foundation—**OneLake**—while using tools designed for their responsibilities.

Key technologies:

### **Enable and use Microsoft Fabric**

#### 1. Enabling Microsoft Fabric
Fabric must be enabled before users can access its capabilities.
##### Roles That Can Enable Fabric
* **Fabric Administrator** – Manages Fabric settings and configurations.
* **Power Platform Administrator** – Administers Power Platform services, including Fabric.
* **Global Administrator** – Has Fabric administration rights through organization-wide permissions.
##### Where to Enable Fabric
> Power BI Service → Admin Portal → Tenant Settings

Administrators can:
* Enable Fabric for the entire organization.
* Enable Fabric for specific Microsoft 365 or Microsoft Entra security groups.
* Delegate enablement at the capacity level.
#### 2. Fabric Workspaces
A workspace is the primary collaboration environment in Fabric.
##### Workspaces contain:
* Lakehouses
* Warehouses
* Reports
* Pipelines
* Notebooks
* Other Fabric items
##### Workspace Features
* All workspace data is stored in **OneLake**.
* Supports **Data Lineage** View to visualize dependencies and data flow.
* Enables team collaboration and resource organization.
##### Workspace Settings
You can configure:
* License type
* OneDrive access
* Azure Data Lake Gen2 connectivity
* Git integration
* Spark settings
##### Workspace Roles
| Role | Purpose |
| --- | --- |
| Admin | Full control |
| Member | Create and manage content |
| Contributor | Create and edit content |
| Viewer | Read-only access |

**Best Practice**: Use workspace roles for collaboration and item-level permissions for finer access control.

#### 3. Discovering Data with OneLake Catalog
The **OneLake Catalog** helps users find data assets across the organization.
##### Capabilities:
* Browse shared content.
* Filter by workspace.
* Filter by domain.
* Search by keyword.
* Filter by item type.
##### Important Point
Users only see items that have been shared with them.

#### 6. Fabric IQ (Preview)
**Fabric IQ** is an AI-focused Fabric workload.
##### Purpose
Organize organizational data according to business language instead of raw databases and table structures.
##### Core Component: Ontology
An ontology defines:
* Business concepts
* Relationships
* Rules

This allows AI systems to reason using business terminology.
#### 7. The Three Microsoft IQ Workloads
##### Fabric IQ
Focuses on:
* OneLake
* Semantic models
* Business graphs
* Analytics data

**Purpose**: Enable AI to understand business analytics.
##### Foundry IQ
Connects:
* Azure
* SharePoint
* OneLake
* Web content

**Purpose**: Provide permission-aware enterprise knowledge.
##### Work IQ
Uses:
* Documents
* Meetings
* Chats
* Workflows

**Purpose**: Help AI understand how an organization works.
##### Easy Exam Memory Aid
| IQ Type | Focus |
| --- | --- |
| Fabric IQ | Business data |
| Foundry IQ | Enterprise knowledge |
| Work IQ | Work activities |
#### 8. Fabric Data Agents
**Fabric Data Agents** provide conversational access to data.
##### Users can:
* Ask questions in natural language.
* Query lakehouses.
* Query warehouses.
* Query semantic models.
##### When connected to Fabric IQ, agents can:
* Understand business concepts defined in the ontology.
* Return more meaningful answers.
#### 9. Copilot in Microsoft Fabric
Copilot is available across Fabric workloads.
##### Code Assistance
Copilot can:
* Generate code
* Complete code
* Generate SQL from natural language
* Generate KQL queries
##### Data Transformation Assistance
In Data Factory, Copilot helps:
* Create transformations
* Explain transformation logic
* Support both technical and low-code users
##### Reporting Assistance
In Power BI, Copilot can:
* Generate reports
* Create page summaries
* Answer questions about data using natural language
#### Important Exam Facts
* Fabric is enabled from Admin Portal → Tenant Settings
* OneLake stores all Fabric data
* Workspaces are the main collaboration containers
* Workspace roles: Admin, Member, Contributor, Viewer
* OneLake Catalog helps discover shared data assets
* Fabric IQ uses ontologies to model business concepts
* Three IQ workloads: Fabric IQ, Foundry IQ, Work IQ
* Data Agents allow natural-language querying of organizational data
* Copilot supports code generation, transformations, and report creation
* Copilot in Fabric is enabled by default and can be controlled through Tenant Settings
#### Quick Memory Formula
> OneLake → Workspaces → Fabric Workloads → Fabric IQ → Data Agents → Copilot

This is the flow from storing data, to managing it, to enabling AI-powered analytics and business insights.
